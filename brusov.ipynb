```python
import pandas as pd
import spacy
from spacy.lang.ru.examples import sentences 

nlp = spacy.load("ru_core_news_sm")

with open('brusov.txt', 'r', encoding='Windows-1251') as file:
    text = file.read()

doc = nlp(text)   
tokens_data = []

for token in doc:
    tokens_data.append({
        'Токен': token.text,
        'Лемма': token.lemma_,
        'Часть речи': token.pos_})
df = pd.DataFrame(tokens_data)

output_file = 'brusov.data.spacy.xlsx'
df.to_excel(output_file, index = False)
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag
import pandas as pd

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')

with open('brusov.txt', 'r', encoding='Windows-1251') as file:
    text = file.read()
    
tokens = word_tokenize(text)
lemmatizer = WordNetLemmatizer()

token_list = []
lemma_list = []
pos_list = []

for token in tokens:
    token_list.append(token)
    parsed_token = morph.parse(token)[0]
    lemma = parsed_token.normal_form
    lemma_list.append(lemma)

    pos = parsed_token.tag.POS
    pos_list.append(pos)

df = pd.DataFrame({
    'Token': token_list,
    'Lemma': lemma_list,
    'Part of Speech': pos_list
})

df.to_excel('brusov.data.nltk.xlsx', index=False, engine='openpyxl')

if token_list: 
    print(token_list[-1])  

```

```python
import spacy
from spacy.lang.en.examples import sentences 
nlp = spacy.load('en_core_web_sm')
import pandas as pd

with open('tex.txt', 'r', encoding='utf-8') as file:
    text = file.read()
doc = nlp(text)
data = []
for token in doc:
    data.append({
        "Text": token.text,
        "Lemma": token.lemma_,
        "POS": token.pos_,
        "Tag": token.tag_,
    })

df = pd.DataFrame(data)

output_file = "tex.morph.xlsx"
df.to_excel(output_file, index=False)
```
