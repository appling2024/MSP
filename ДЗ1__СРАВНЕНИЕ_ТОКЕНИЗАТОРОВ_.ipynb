```
with open('mazay.txt', 'r', encoding='utf-8') as file:
    text = file.read()


import spacy
nlp = spacy.load("ru_core_news_sm")
doc = nlp(text)
tokens_spacy = [token.text for token in doc]


import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')
tokens_nltk = word_tokenize(text, language='russian')


from razdel import tokenize
tokens_razdel = [token.text for token in tokenize(text)]
with open('tokens_spacy.txt', 'w', encoding='utf-8') as f:
    f.write('\n'.join(tokens_spacy))

with open('tokens_nltk.txt', 'w', encoding='utf-8') as f:
    f.write('\n'.join(tokens_nltk))

with open('tokens_razdel.txt', 'w', encoding='utf-8') as f:
    f.write('\n'.join(tokens_razdel))


sets = {
    'spaCy': set(tokens_spacy),
    'NLTK': set(tokens_nltk),
    'Razdel': set(tokens_razdel)
}

all_tokens = set.union(*sets.values())
differences = {lib: all_tokens - tokens for lib, tokens in sets.items()}

for lib, diff in differences.items():
    print(f"Tokens not found in {lib}:")
    print(diff)
```
