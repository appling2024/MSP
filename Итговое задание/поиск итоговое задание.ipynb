```python
import os
import xml.etree.ElementTree as ET

# Папка с файлами из корпуса CORPORA
folder_path = "C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/Итговое задание/annot.opcorpora.xml.byfile"

def search_in_opcorpora(folder_path, lemma=None, morph_tag=None):
    results = []
    for file_name in os.listdir(folder_path):
        file_path = os.path.join(folder_path, file_name)
        if file_name.endswith(".xml"):
            try:
                print(f"Читаем файл: {file_name}")
                tree = ET.parse(file_path)
                root = tree.getroot()

                for token in root.iter("token"):
                    token_lemma = None
                    token_morph_tags = []

                    # Извлекаем лемму
                    for l in token.iter("l"):
                        token_lemma = l.attrib.get("t")

                    # Извлекаем морфологические теги
                    for g in token.iter("g"):
                        token_morph_tags.append(g.attrib.get("v"))

                    # Условия для поиска
                    if lemma and token_lemma == lemma:
                        results.append({
                            "file": file_name,
                            "lemma": token_lemma,
                            "morph_tags": token_morph_tags
                        })
                    elif morph_tag and morph_tag in token_morph_tags:
                        results.append({
                            "file": file_name,
                            "lemma": token_lemma,
                            "morph_tags": token_morph_tags
                        })
            except Exception as e:
                print(f"Ошибка при обработке файла {file_name}: {e}")

    print("Результаты поиска:")
    for result in results:
        print(f"Файл: {result['file']}, Лемма: {result['lemma']}, Морфологические теги: {', '.join(result['morph_tags'])}")

# поиск только по лемме
search_in_opcorpora(
    folder_path=folder_path,
    lemma="холодный",
    morph_tag=None   
)

# поиск только по морфологическому тегу
search_in_opcorpora(
    folder_path=folder_path,
    lemma=None,   
    morph_tag="VERB"   
)
```
