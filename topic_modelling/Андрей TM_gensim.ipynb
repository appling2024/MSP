```python
import codecs
import os
```

```python
path = r'annot' # TODO вставить путь до директории с предобработанным корпусом
file_names = os.listdir(path) 
```

```python
data = []
for name in file_names:
    with codecs.open(path + "/" + name, encoding = 'utf-8') as f:
        data.append(f.read())
```

```python
from gensim import corpora, models
```

```python
# Костыль
documents = [text.split() for text in data]
```

```python
from gensim import corpora

# Map each token to a unique ID
dictionary = corpora.Dictionary(documents)
print(f'Number of unique tokens: {len(dictionary)}')

# Filter out tokens by frequency
min_doc, max_doc = 15, .2
dictionary.filter_extremes(no_below=min_doc, no_above=max_doc)
print(f'Number of unique tokens (filtered): {len(dictionary)}')

# Create a BOW corpus
corpus = [dictionary.doc2bow(doc) for doc in documents]
print(f'Number of documents: {len(corpus)}')
```

```python
NUM_TOPICS = 10
```

```python
more_stop_words=['язык','например','сантиметр']
```

```python
tokens_in_documents = []
for document in data:
    tokens_in_this_one = [token for token in document.split(" ") if token not in more_stop_words]
    tokens_in_documents.append(tokens_in_this_one)
```

```python
tokens_in_documents[0][:10]
```

```python
# Build a Dictionary - association word to numeric id
dictionary = corpora.Dictionary(tokens_in_documents)
```

```python
# Transform the collection of texts to a numerical form
corpus = [dictionary.doc2bow(text) for text in tokens_in_documents]
```

```python
# Have a look at how the first document looks like: [(word_id, count), ...]
print(corpus[0])
```

```python
# Build the LDA model
lda_model = models.LdaModel(corpus=corpus, num_topics=NUM_TOPICS, id2word=dictionary, random_state=3)
 
# Build the LSI model
lsi_model = models.LsiModel(corpus=corpus, num_topics=NUM_TOPICS, id2word=dictionary)
```

```python
# display the topics

print("LDA Model:")
 
for idx in range(NUM_TOPICS):
    # Print the first 10 most representative topics
    print("Topic #%s:" % idx, lda_model.print_topic(idx, 10))

print("=" * 20)
print("LSI Model:")
 
for idx in range(NUM_TOPICS):
    # Print the first 10 most representative topics
    print("Topic #%s:" % idx, lsi_model.print_topic(idx, 10))

print("=" * 20)
```
