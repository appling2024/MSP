# Разработать программу для автоматической оценки качества морфологической разметки

```
import pandas as pd
```

```
# читать данные из файлов
def read_tagged_file(file_path):
    tagged_data = []
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            words_tags = line.strip().split()
            if len(words_tags) == 2:
                word, tag = words_tags
                tagged_data.append((word, tag))
    return tagged_data
```

```
# сравнение результатов классификаций
def evaluate_tagging(gold_standard, tagged_output):
    correct = 0
    total = len(gold_standard)
    errors = []

    for gs, output in zip(gold_standard, tagged_output):
        if gs[1] == output[1]:
            correct += 1
        else:
            errors.append((gs[0], gs[1], output[1]))

    accuracy = (correct / total) * 100
    return accuracy, errors
```

```
# вывод результатов в упорядоченном виде
def print_results(accuracy, errors):
    print(f"точность: {accuracy:.2f}%")
    if errors:
        print("\nОшибки: ")
        for error in errors:
            print(f"Слово: {error[0]}, правильная классификация : {error[1]}, итоговая классификация : {error[2]}")
    else:
        print("\nОшибок в кклассификации нет.")
```

```
if __name__ == "__main__":
    # определение путей к файлам
    gold_standard_file = 'gold_standard.txt'  # الملف المعياري
    tagged_output_file = 'tagged_output.txt'  # الملف الناتج من التاغر

    # чтение файлов
    gold_standard = read_tagged_file(gold_standard_file)
    tagged_output = read_tagged_file(tagged_output_file)

    # оценка классификации
    accuracy, errors = evaluate_tagging(gold_standard, tagged_output)

    print_results(accuracy, errors)
```
