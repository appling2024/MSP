```python
import pandas as pd

# Загрузка таблиц
df1 = pd.read_csv("/Users/elizavetakadochnikova/Desktop/keywords_comparison.csv")
df2 = pd.read_csv("/Users/elizavetakadochnikova/Desktop/comparison_results-2.csv")

# Переименование столбцов
df1_renamed = df1.rename(columns={
    "Generated keywords": "Generated 1",
    "Reference keywords": "Reference keywords"
})

df2_renamed = df2.rename(columns={
    "Generated keywords": "Generated 2"
})

# Объединение таблиц по Article ID
merged_df = pd.merge(
    df1_renamed[['Article ID', 'Generated 1', 'Reference keywords']],
    df2_renamed[['Article ID', 'Generated 2']],
    on='Article ID',
    how='outer'
)

# Функция для подсчёта совпадений
def calculate_match_percent(generated, reference):
    if pd.isna(generated) or pd.isna(reference):
        return None
    gen_set = set([kw.strip().lower() for kw in generated.split(',') if kw.strip()])
    ref_set = set([kw.strip().lower() for kw in reference.split(',') if kw.strip()])
    if not ref_set:
        return 0
    return round(100 * len(gen_set & ref_set) / len(ref_set), 2)

# Вычисление совпадений
merged_df["Match % (Generated 1)"] = merged_df.apply(
    lambda row: calculate_match_percent(row["Generated 1"], row["Reference keywords"]), axis=1
)

merged_df["Match % (Generated 2)"] = merged_df.apply(
    lambda row: calculate_match_percent(row["Generated 2"], row["Reference keywords"]), axis=1
)

# Сохранение результата
merged_df.to_csv("KW_DONE.csv", index=False)
```
