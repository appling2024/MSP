```python
import langid
from langid.langid import LanguageIdentifier, model
identifier = LanguageIdentifier.from_modelstring(model, norm_probs=True)
identifier.classify('this is a test')
identifier.set_languages(['en','it'])
identifier.classify('this is a test')
```

```python
import langdetect
from langdetect import detect_langs
detect_langs('this is a test')
```

```python
import csv
from langdetect import detect_langs

with open('test.csv', 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    i = 1

    for row in reader:
        text = row[5] 
        languages = detect_langs(text) 
        for lang in languages:
          lang_code = lang.lang  
          file_name = f"file_{lang_code}.txt" 
          with open(file_name, 'a', encoding='utf-8') as lang_file:
                        lang_file.write(f"Row {i}:\n{text}\n\n") 
        i += 1

```
