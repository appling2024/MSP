```python
import random

class House:
    def __init__(self):
        self.money = 20
        self.food = 1
        self.dirt = 0

    def __str__(self):
        return f'В доме {self.money} денег, {self.food} еды, грязи: {self.dirt}'

class Human:
    total_shop_count = 0

    def __init__(self, name, house):
        self.name = name
        self.fullness = 50
        self.house = house
        self.shop_count = 0

    def __str__(self):
        return (f'Я {self.name}! Моя сытость {self.fullness}, '
                f'в доме {self.house.food} еды и {self.house.money} денег, '
                f'я ходил в магазин {self.shop_count} раз(а).')

    def work(self):
        self.fullness -= 10
        self.house.money += 10
        print(f'{self.name} поработал')

    def shop(self):
        if self.house.money >= 10:
            self.house.money -= 10
            self.house.food += 5
            self.shop_count += 1
            Human.total_shop_count += 1
            print(f'{self.name} сходил в магазин')

    def eat(self):
        if self.house.food >= 2:
            self.fullness += 10
            self.house.food -= 2
            print(f'{self.name} поел')

    def clean_house(self):
        if self.house.dirt >= 30:
            self.house.dirt -= 30
            self.fullness -= 10
            print(f'{self.name} убрался в доме')

    def watch_MTV(self):
        self.fullness -= 5
        print(f'{self.name} посмотрел телевизор')

    def act(self):
        if self.fullness <= 0:
            print(f'{self.name} умер от голода...')
            return False
        elif self.fullness < 20:
            self.eat()
        elif self.house.food < 10:
            self.shop()
        elif self.house.money < 10:
            self.work()
        elif self.house.dirt >= 30:
            self.clean_house()
        else:
            dice = random.randint(1, 6)
            if dice == 1:
                self.watch_MTV()
            elif dice in (2, 3, 4):
                self.work()
            else:
                self.eat()
        return True

class Cat:
    def __init__(self, name, house):
        self.name = name
        self.fullness = 30
        self.house = house

    def __str__(self):
        return f'Кот {self.name}, сытость {self.fullness}'

    def eat(self):
        if self.house.food >= 2:
            self.fullness += 10
            self.house.food -= 2
            print(f'Кот {self.name} поел')

    def sleep(self):
        self.fullness -= 5
        print(f'Кот {self.name} спит')

    def scratch_wallpaper(self):
        self.fullness -= 10
        self.house.dirt += 5
        print(f'Кот {self.name} дерет обои')

    def act(self):
        if self.fullness <= 0:
            print(f'Кот {self.name} ушел к новым хозяевам из-за голода...')
            return False
        elif self.fullness < 20:
            self.eat()
        else:
            dice = random.randint(1, 2)
            if dice == 1:
                self.sleep()
            else:
                self.scratch_wallpaper()
        return True

berloga = House()
vasya = Human('Вася', berloga)
petya = Human('Петя', berloga)
yura = Human('Юра', berloga)
barsik = Cat('Барсик', berloga)
murzik = Cat('Мурзик', berloga)
pushok = Cat('Пушок', berloga)

humans = [vasya, petya, yura]

for day in range(1, 366):
    print(f'======================== День {day} ========================')

    random.shuffle(humans)

    for human in humans:
        if human.fullness < 20:
            human.eat()
        elif human.house.money >= 10:
            human.shop()

    if not vasya.act() or not petya.act() or not yura.act() or not barsik.act() or not murzik.act() or not pushok.act():
        break

    print(berloga)
    print(vasya)
    print(petya)
    print(yura)
    print(barsik)
    print(murzik)
    print(pushok)
    print("\n")

print(f'Общее количество походов в магазин за год: {Human.total_shop_count}')
```

```python
import random

class House:
    def __init__(self):
        self.money = 150
        self.food = 20
        self.dirt = 0

    def __str__(self):
        return f'В доме {self.money} денег, {self.food} еды, грязи: {self.dirt}'

class Human:
    total_shop_count = 0

    def __init__(self, name, house):
        self.name = name
        self.fullness = 50
        self.house = house
        self.shop_count = 0

    def __str__(self):
        return (f'Я {self.name}! Моя сытость {self.fullness}, '
                f'в доме {self.house.food} еды и {self.house.money} денег, '
                f'я ходил в магазин {self.shop_count} раз(а).')

    def work(self):
        self.fullness -= 3
        self.house.money += 50
        print(f'{self.name} поработал')

    def shop(self):
        if self.house.money >= 30:
            self.house.money -= 30
            food_to_add = 15
            self.house.food += food_to_add
            self.shop_count += 1
            Human.total_shop_count += 1
            print(f'{self.name} сходил в магазин и купил {food_to_add} еды')

    def eat(self):
        if self.house.food >= 5:
            self.fullness += 20
            self.house.food -= 5
            print(f'{self.name} поел')

    def clean_house(self):
        if self.house.dirt >= 20:
            self.house.dirt -= 20
            self.fullness -= 3
            print(f'{self.name} убрался в доме')

    def watch_MTV(self):
        self.fullness -= 3
        print(f'{self.name} посмотрел телевизор')

    def act(self):
        if self.fullness <= 0:
            print(f'{self.name} умер от голода...')
            return False

        dice = random.randint(1, 6)

        if dice == 1:
            self.watch_MTV()
        elif dice == 2:
            self.work()
        elif dice == 4:
            self.eat()
        elif dice in (3, 5):
            self.shop()
        elif dice == 6 and self.house.dirt >= 20:
            self.clean_house()

        if self.fullness <= 0:
            print(f'{self.name} умер от голода...')
            return False

        return True

class Cat:
    def __init__(self, name, house):
        self.name = name
        self.fullness = 30
        self.house = house

    def __str__(self):
        return f'Кот {self.name}, сытость {self.fullness}'

    def eat(self):
        if self.house.food >= 2:
            self.fullness += 10
            self.house.food -= 2
            print(f'Кот {self.name} поел')

    def sleep(self):
        self.fullness -= 3
        print(f'Кот {self.name} спит')

    def scratch_wallpaper(self):
        self.fullness -= 5
        self.house.dirt += 5
        print(f'Кот {self.name} дерет обои. Грязи в доме: {self.house.dirt}')

    def act(self):
        if self.fullness <= 0:
            print(f'Кот {self.name} ушел к новым хозяевам из-за голода...')
            return False

        dice = random.randint(1, 6)

        if dice in (1, 2):
            if self.house.food >= 2:
                self.eat()
            else:
                print(f'Кот {self.name} не может поесть, еды недостаточно')
        elif dice in (3, 4):
            self.sleep()
        else:
            self.scratch_wallpaper()

        if self.fullness <= 0:
            print(f'Кот {self.name} ушел к новым хозяевам из-за голода...')
            return False

        return True

berloga = House()
vasya = Human('Вася', berloga)
petya = Human('Петя', berloga)
yura = Human('Юра', berloga)
barsik = Cat('Барсик', berloga)
murzik = Cat('Мурзик', berloga)
pushok = Cat('Пушок', berloga)

humans = [vasya, petya, yura]

for day in range(1, 366):
    print(f'======================== День {day} ========================')

    if not vasya.act() or not petya.act() or not yura.act() or not barsik.act() or not murzik.act() or not pushok.act():
        break

    print(berloga)
    print(vasya)
    print(petya)
    print(yura)
    print(barsik)
    print(murzik)
    print(pushok)
    print("\n")

print(f'Общее количество походов в магазин за год: {Human.total_shop_count}')
```
