```python
from collections import Counter
from datetime import datetime

districts = Counter()
crime_types = Counter()
time_counts = Counter()
time_of_day_counts = Counter()
day_of_week_counts = Counter()

with open('SacramentoCrimeJanuary2006.csv', 'r') as all_crimes:
    all_crimes.readline()

    for line in all_crimes:
        items = line.split(',')
        cdatetime = items[0]
        district = items[2]
        crimedescr = items[5]
        districts[district] += 1
        crime_types[crimedescr] += 1

        datetime_obj = datetime.strptime(cdatetime, '%m/%d/%y %H:%M')
        full_datetime = datetime_obj.strftime('%Y-%m-%d %H:%M')
        time_counts[full_datetime] += 1
        time_of_day = datetime_obj.strftime('%H:%M')
        time_of_day_counts[time_of_day] += 1
        day_of_week = datetime_obj.strftime('%A')
        day_of_week_counts[day_of_week] += 1

print("Полицейский округ с наибольшим числом преступлений:", districts.most_common(1))
print("10 самых частых преступлений:")
for crime, count in crime_types.most_common(10):
    print(crime, count)
print("30 самых опасных моментов времени:")
for moment, count in time_counts.most_common(30):
    print(moment, count)
print("10 самых опасных моментов суток:")
for time, count in time_of_day_counts.most_common(10):
    print(time, count)
print("Число преступлений по дням недели:")
for day, count in day_of_week_counts.items():
    print(day, count)
```
