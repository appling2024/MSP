```python
import random

class Human:
    '''Человечище'''
    def __init__(self, name):
        self.name = name
        self.fullness = 50
        self.food = 30
        self.money = 10

    def __str__(self):
        return f'Я {self.name}! Моя сытость {self.fullness}, у меня осталось {self.food} еды и {self.money} денег.'

    def work(self):
        self.fullness -= 10
        self.money += 20
        print(f'{self.name} поработал')

    def shop(self):
        if self.money >= 10:
            self.money -= 10
            self.food += 20
            print(f'{self.name} сходил в магазин')
        else:
            print(f'{self.name} недостаточно денег для похода в магазин')

    def eat(self):
        if self.food >= 10:
            self.fullness += 10
            self.food -= 10    
            print(f'{self.name} поел')
        else:
            print(f'{self.name} недостаточно еды для приёма пищи')

    def watch_MTV(self):
        self.fullness -= 5
        print(f'{self.name} посмотрел телевизор')

    def help_friend(self, friend):
        if self.food >= 10 and friend.fullness < 20:
            friend.fullness += 10
            self.food -= 10
            print(f'{self.name} поделился едой с {friend.name}')

        if self.money >= 5 and friend.money < 5:
            friend.money += 5
            self.money -= 5
            print(f'{self.name} поделился деньгами с {friend.name}')

    def act(self, friends):
        if self.money < 5:
            self.fullness -= 3
            print(f'{self.name} страдает от нехватки денег, его сытость снижается!')
        
        if self.fullness <= 0:
            print(f'{self.name} умер от голода...')
            return False
        elif self.fullness < 20:
            self.eat()
        elif self.food < 10:
            self.shop()
        elif self.money < 20:
            self.work()
        else:
            for friend in friends:
                self.help_friend(friend)
            
            dice = random.randint(1, 6)
            if dice == 1:
                self.watch_MTV()
            elif dice in (2, 3):
                self.work()
            else:
                self.eat()
        return True


vasya = Human('Вася')
petya = Human('Петя')
yura = Human('Юра')

for day in range(1, 366):
    print(f'======================== День {day} ========================')
    if not vasya.act([petya, yura]) or not petya.act([vasya, yura]) or not yura.act([vasya, petya]):
        break  
    print(vasya)
    print(petya)
    print(yura)
    print("\n")
```
