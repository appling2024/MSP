```python
class Backpack:
    """ Рюкзак """

    def __init__(self, gift=None):
        self.content = []
        # if gift:
        #     self.content.append(gift)

    def __str__(self):
        return 'Backpack: '+' '.join(self.content)
    
    def add(self, item):
        """ Положить в рюкзак """
        self.content.append(item)
        print('В рюкзак положили:', item)

    def __add__(self, other):
        new_obj = Backpack()
        new_obj.content.extend(self.content)
        new_obj.content.extend(other.content)
        return new_obj
    
    def __len__(self):
        return len(self.content)
    
    def __bool__(self):
        return self.content != []


my_backpack = Backpack()
my_backpack.add('яблоко')
her_backpack = Backpack(gift = 'гиря 2кг')
her_backpack.add('пистолет')
new_backpack = her_backpack + my_backpack

print(new_backpack)
print(my_backpack)
print(her_backpack)

# list1 = ['Маша', 'Лена', 'Таня']  
# list_string = ', '.join(list1)
# print(list_string)  
```

```python
class Crew:
    def __init__(self, members=set()):
        self.members = members
    
    def __str__(self):
        return 'Состав команды: '+', '.join(self.members)
    
    def __len__(self):
        return len(self.members)
    
    def __add__(self, other):
        return Crew(self.members | other.members)
    
    def __sub__(self, other):
        return Crew(self.members - other.members)
    
crew1 = Crew({'Маша', 'Петя', 'Вася'})
crew2 = Crew({'Петя', 'Юра'})

crew3 = crew1 + crew2
crew4 = crew1 - crew2

print(crew1)
print(f"Количество участников в crew1: {len(crew1)}")
print(crew2)
print(f"Количество участников в crew2: {len(crew2)}")
print(crew3)
print(f"Количество участников в crew3: {len(crew3)}")
print(crew4)
print(f"Количество участников в crew4: {len(crew4)}")
```
