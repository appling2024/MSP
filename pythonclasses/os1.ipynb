```python
import time
import os

path = 'C:/Windows/help'
path_normalized = os.path.normpath(path)
print(path_normalized)

count = 0
for dirpath, dirnames, filenames in os.walk(path_normalized):
    print('*' * 27)
    print(dirpath, dirnames, filenames)
    print(os.path.dirname(dirpath))
    count += len(filenames)
    for file in filenames:
        full_file_path = os.path.join(dirpath, file)
        secnds = os.path.getmtime(full_file_path)
        file_time = time.gmtime(secnds)
        if file_time[0] == 2022:  # год можно указать другой
            # выводим только файлы за 2022 год
            print(full_file_path, secnds, file_time)
print(count)


```

```python
import os
import shutil

directory_path = 'C:/Users/user/Documents/GitHub/MSP/pythonclasses/os_homework'

image_files = []
empty_folders = []

for dirpath, dirnames, filenames in os.walk(directory_path):
    for filename in filenames:
        if filename.endswith('.jpg'):
            full_path = os.path.join(dirpath, filename)
            image_files.append(full_path)

    if not dirnames and not filenames:
        empty_folders.append(dirpath)

print("Список картинок:")
for image in image_files:
    print(image)

print("\nСписок пустых папок:")
for folder in empty_folders:
    print(folder)

pictures_folder = 'C:/Users/user/Documents/GitHub/MSP/pythonclasses/os_homework/pictures'
os.makedirs(pictures_folder, exist_ok=True)

for image in image_files:
    shutil.copy2(image, pictures_folder)
```
