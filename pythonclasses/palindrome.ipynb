```python
lst = [0, 15, 15, 3]
def remove_duplicates(lst):
    set_lst = set(lst)
    lst2 = list(set_lst)
    return lst2
list1 = [1,5,1,8,8,1,6]
list2 = remove_duplicates(list1)
print(list2)
```

```python
def is_palindrome(lst):
    if lst == lst.reverse():
        return True
    else:
        return False
list1 = [1, 5, 1, 8, 8, 1, 6]
list2 = is_palindrome(list1)
print(list2)
```

```python
def rotate_list(lst, n):
    return lst[n:] + lst[:n]

list1 = [1, 5, 1, 8, 8, 1, 6]
n = 3  
list2 = rotate_list(list1, n)
print(list2)
```

```python
# 8. Если при распаковке количество переменных слева больше, чем количество элементов в кортеже:
#    Python выдаст ошибку `ValueError: not enough values to unpack`. Это происходит потому, что Python ожидает, что количество переменных слева совпадёт с количеством элементов в кортеже, и если их меньше, не сможет заполнить все переменные значениями.

b = (1, 2)
x, y, z = b  # Ошибка: ValueError: not enough values to unpack (expected 3, got 2)

# 9. Если при распаковке количество переменных слева меньше, чем количество элементов в кортеже:
#    Python также выдаст ошибку `ValueError: too many values to unpack`. Это происходит, так как остаются элементы, для которых нет переменных, чтобы принять их значения.

b = (1, 2, 3)
x, y = b  # Ошибка: ValueError: too many values to unpack (expected 2)

# 10. Можно ли аналогичным образом распаковать список (если число элементов в нем совпадает с числом переменных)?
#     Да, Python позволяет распаковывать списки точно так же, как кортежи, если количество элементов в списке соответствует количеству переменных слева. Списки и кортежи в этом случае работают одинаково.

test_list = [1, 2, 3]
x, y, z = test_list  # Всё работает корректно
print(x, y, z)  # Вывод: 1 2 3

# 11. Что будет, если распаковать в три переменные слово 'pig'? Можно ли вообще распаковывать строки?
#    Да, строки можно распаковывать аналогично спискам и кортежам, так как строка — это итерируемый объект. При распаковке строки 'pig' в три переменные каждая переменная получит по одному символу из строки 'p', 'i' и 'g'. Однако, как и в случае с кортежами и списками, количество переменных должно совпадать с количеством символов.

word = 'pig'
x, y, z = word  # Корректно
print(x, y, z)  # Вывод: p i g
```

```python
def merge_tuples(tuple1, tuple2):
    return tuple(set(tuple1) | set(tuple2))

tuple1 = (1, 2, 3, 4, 5)
tuple2 = (4, 5, 6, 7, 8)
result = merge_tuples(tuple1, tuple2)
print(result) 
```

```python
def merge_tuples(tuple1, tuple2):
    unique_elements = list(tuple1)
    for item in tuple2:
        if item not in unique_elements:
            unique_elements.append(item)
    return tuple(unique_elements)

tuple1 = (1, 2, 3, 4, 5)
tuple2 = (4, 5, 6, 7, 8)
result = merge_tuples(tuple1, tuple2)
print(result) 
```

```python
def merge_tuples(tuple1, tuple2):
    tup_tmp = tuple1 + tuple2
    set1 = set(tup_tmp)
    return tuple(set1)

tuple1 = (1, 2, 3, 4, 5)
tuple2 = (4, 5, 6, 7, 8)
result = merge_tuples(tuple1, tuple2)
print(result) 
```
