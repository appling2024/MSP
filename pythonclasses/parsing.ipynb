```python
import requests
from bs4 import BeautifulSoup

r = requests.get('https://www.interfax.ru/rss.asp')

soup = BeautifulSoup(r.content, features='html')
items = soup.find_all('item')

with open('all_articles.txt', 'w', encoding='utf-8') as file:
    for item in items:
        title = item.find('title').text
        link = item.find('guid').text
        
        file.write(f"Заголовок: {title}\n")
        file.write(f"Ссылка: {link}\n")
        
        article_r = requests.get(link)
        
        if article_r.status_code == 200:
            article_soup = BeautifulSoup(article_r.content, 'html.parser')

            content = article_soup.find('article', itemprop='articleBody')
            if content:
                paragraphs = content.find_all('p')

                file.write("Текст статьи:\n")
                for p in paragraphs:
                    file.write(f"{p.text.strip()}\n")
                
                file.write("\n" + "-"*50 + "\n\n")
                print(f"Содержимое статьи '{title}' добавлено в файл.")
            else:
                file.write("Не удалось найти основной текст статьи.\n")
                file.write("\n" + "-"*50 + "\n\n")
        else:
            file.write(f"Не удалось загрузить статью по ссылке: {link}\n")
            file.write("\n" + "-"*50 + "\n\n")

print("Все данные сохранены в файл 'all_articles.txt'.")
```
