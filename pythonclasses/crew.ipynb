```python
crew = []
while True:
    name = input('Имя: ')
    print('Привет! ', name)
    crew.append(name)
    if name == 'Паганель':
        break
crew.sort()
for name in crew:
    print('Наша команда: ', name)
```

```python
my_file1 = open('makulatura.txt', 'r')

total_makulatura = 0
num_people = 0

my_file2 = open('letters.txt', 'w')

print('\tПЕРЕДОВИКИ (сдали больше или равно 25 кг.):')

for line in my_file1:
    items = line.split('-')
    name = items[0].strip()
    weight = int(items[1])  

    total_makulatura += weight
    num_people += 1

    if weight >= 25:
        print(name, ':', weight, 'кг.')
    elif weight < 20:
        debt = 20 - weight
        letter = ('Уважаемый(ая) ' + name + '!\n' + 
                  'Напоминаем, что Ваш долг по сдаче макулатуры составляет ' + str(debt) + ' кг.\n' +
                  'Просьба сдать недостающую макулатуру до 17 октября.\n')
        if debt > 10:
            letter += 'Иначе Вас ждет суровое наказание!\n'
        letter += 'Администрация\n'
        letter += '----------------------\n'
        my_file2.write(letter)

my_file1.close()

print('ВСЕГО СДАЛИ:', total_makulatura, 'кг.')
print('СРЕДНЕЕ КОЛИЧЕСТВО МАКУЛАТУРЫ НА ЧЕЛОВЕКА: ', total_makulatura / num_people, 'кг.')

my_file2.close()
```
