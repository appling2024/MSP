```python
import re
from collections import Counter
from nltk import pos_tag, word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
import nltk
```

```python
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
```

```python
def analyze_file(file_name):
    """Открывает файл и выполняет анализ текста"""
    
    file_name = "f1.txt"  #<------ сюда записывать название файла
    with open(file_name, 'r', encoding='utf-8') as file:
        text = file.read()

    tokens = word_tokenize(text)

    lemmatizer = WordNetLemmatizer()
    lemmas = [lemmatizer.lemmatize(token.lower()) for token in tokens]

    pos_tags = pos_tag(tokens)

    nouns = [word for word, pos in pos_tags if pos.startswith('NN')]

    num_tokens = len(tokens)
    num_nouns = len(nouns)
    noun_counts = Counter(nouns).most_common(5)

    lemmas_starting_with_a = [lemma for lemma in lemmas if lemma.startswith('a')]
    lemma_a_counts = Counter(lemmas_starting_with_a).most_common(5)

    first_letters = [lemma[0] for lemma in lemmas if lemma.isalpha()]
    first_letter_counts = Counter(first_letters).most_common(5)

    results = {
        'Количество токенов': num_tokens,
        'Количество существительных': num_nouns,
        'Пять самых частых существительных и их количество': noun_counts,
        'Пять самых частых лемм, начинающихся на букву "а", и их количество': lemma_a_counts,
        'Пять самых частых букв, с которых начинаются леммы': first_letter_counts,
    }

    print("Результаты анализа файла:")
    for key, value in results.items():
        print(f"{key}: {value}")

analyze_file(file_name)
```

```python
class Texter:
    def __init__(self, file_name):
        """Инициализирует объект с текстом из файла"""
        file_name = "f1.txt"
        with open(file_name, 'r', encoding='utf-8') as file:
            self.text = file.read()

        self.tokens = word_tokenize(self.text)
        self.lemmatizer = WordNetLemmatizer()
        self.lemmas = [self.lemmatizer.lemmatize(token.lower()) for token in self.tokens]
        self.pos_tags = pos_tag(self.tokens)

    def __str__(self):
        """Возвращает текст объекта"""
        return self.text

    def get_num_tokens(self):
        return len(self.tokens)

    def get_num_nouns(self):
        nouns = [word for word, pos in self.pos_tags if pos.startswith('NN')]
        return len(nouns)

    def get_top_nouns(self):
        nouns = [word for word, pos in self.pos_tags if pos.startswith('NN')]
        return Counter(nouns).most_common(5)

    def get_top_lemmas_a(self):
        lemmas_starting_with_a = [lemma for lemma in self.lemmas if lemma.startswith('a')]
        return Counter(lemmas_starting_with_a).most_common(5)

    def get_top_first_letters(self):
        first_letters = [lemma[0] for lemma in self.lemmas if lemma.isalpha()]
        return Counter(first_letters).most_common(5)


texter = Texter(file_name)
print("Результаты работы методов класса Texter:")
print("Количество токенов:", texter.get_num_tokens())
print("Количество существительных:", texter.get_num_nouns())
print("Пять самых частых существительных:", texter.get_top_nouns())
print("Пять самых частых лемм, начинающихся на 'а':", texter.get_top_lemmas_a())
print("Пять самых частых первых букв лемм:", texter.get_top_first_letters())
```
