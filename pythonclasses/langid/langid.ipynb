```python
import csv
import langid
from langid import classify

with open('test.csv', 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    i = 1

    for row in reader:
        text = row[5]
        lang_code, confidence = classify(text)
        file_name = f"file_{lang_code}.txt"
        with open(file_name, 'a', encoding='utf-8') as lang_file:
            lang_file.write(f"Row {i} (Confidence: {confidence:}):\n{text}\n\n")
        i += 1

```
