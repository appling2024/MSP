```python
plate = 0

while True:
    name = input('Введите имя: ')
    if name == 'Сторож':
            break
    amount = int(input('Введите сумму: '))
    if amount > 0:
        plate += amount
        print('На тарелке: ', amount, 'Спасибо', name, '!')
    elif amount > 0:
         print('Дурной?')
    else:
        if plate + amount >= 0:
            plate += amount
            print('Рады помочь!')
        else:
            print('Таких денег нету') 

print('На тарелке', plate, 'лари')
```
