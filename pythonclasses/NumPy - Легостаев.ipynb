```python
import numpy as np
my_lst = [1, 2, 3]
arr1 = np.array(my_lst)
print(arr1)
arr2 = np.ones((3,10), dtype=int)
print(arr2)
arr3 = np.arange(1,31)
print(arr3)
arr33 = np.arange(0, 16, 5)
print(arr33)
arr34 = np.arange(5)
print(arr34)
arr3 = arr3.reshape(3, 10)
print(arr3)
arr4 = np.arange(1,31).reshape(3, 10)
print(arr4)
arr5 = np.ones((3,10), dtype=int)
arr5 = arr5 * 100
print(arr5)
arr6 = arr4 + arr5
print(arr6)
```

```python
arr3 = np.arange(1,31).reshape(3, 10)
print(arr3)
print("\nПервая строка:")
print(arr3[ 0, :])
print("\nПервый столбец:")
print(arr3[ :, 0])
last_two_columns = arr3[:, -2:]
print("\nДва последних столбца:")
print(last_two_columns)
top_left_square = arr3[:2, :2]
print("\nЛевый верхний квадратик 2×2:")
print(top_left_square)
top_left_square_alt = arr3[:2, :2]
print("\nЛевый верхний квадратик 2×2:")
print(top_left_square_alt)
bottom_right_square = arr3[-2:, -2:]
print("\nПравый нижний квадратик 2×2:")
print(bottom_right_square)
arr3[:, 1::2] = 77 
print("\nМассив после замены чётных столбцов на 77:")
print(arr3)
```

```python
chess = np.zeros((8,8), dtype=int)
chess[1::2, ::2] = 1
chess[::2, 1::2] = 1
print(chess)
```

```python
matrix = np.zeros((10, 10), dtype=int)
for i in range(1, 10):  
    for j in range(i):  
        matrix[i, j] = 5
print(matrix)
```

```python
matrix = np.zeros((10, 10), dtype=int)
number = 1
for i in range(1, 10): 
    for j in range(i): 
        matrix[i, j] = number
        number += 1
print("Матрица с числами ниже диагонали:")
print(matrix)
```

```python
matrix = np.zeros((10, 10), dtype=int)
number = 1
for j in range(10):  
    for i in range(j + 1, 10):  
        matrix[i, j] = number
        number += 1
print("Матрица с числами ниже диагонали (по столбцам):")
print(matrix)
```
