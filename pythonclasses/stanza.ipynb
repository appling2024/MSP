```python
import stanza

# Скачивание и инициализация пайплайна
nlp = stanza.Pipeline('ru', processors='tokenize,pos', logging_level='WARN')

# Тестовый текст
text = "Пример русского текста для обработки."
doc = nlp(text)

# Вывод результатов
for sentence in doc.sentences:
    for word in sentence.words:
        print(f"Слово: {word.text}, POS-тег: {word.upos}")
```
