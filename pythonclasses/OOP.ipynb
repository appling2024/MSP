```python
class cow():
    '''Виртуальная корова'''
    def __init__(self, name, color):
        self.name = name
        self.color = color
        print('Появилась новая корова')
    
    def __str__(self):
        return f'Я корова {self.name}! Мой цвет - {self.color}. МУУУУУУУУУУУУ!'

    def speak(self):
        print('Мууууууу!')

cow1 = cow('Зорька', 'синий')
cow2 = cow('Бурёнка', 'красный')
cow1.speak()
cow2.speak()
print(cow1)
print(cow2)
```
