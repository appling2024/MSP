```python
import random

def get_intent(question):
    question = question.lower()
    if question in ['привет', 'здравствуйте', 'добрый день', 'хелло']:
        intent = 'hello'
    elif question in ['пока', 'до свидания', 'увидимся', 'счастливо']:
        intent = 'bye'
    elif question in ['как дела', 'как ты', 'что нового']:
        intent = 'how_are_you'
    else:
        intent = None
    return intent

def get_sentiment(question):
    sentiment = 0
    return sentiment

def bot(question):
    intent = get_intent(question)
    sentiment = get_sentiment(question)
    
    responses = {
        'hello': ['Привет, человек!', 'Здравствуйте!', 'Хорошего дня!', 'Рад вас видеть!'],
        'bye': ['До свидания!', 'Пока-пока!', 'Увидимся!', 'Счастливого пути!'],
        'how_are_you': ['Всё отлично, спасибо!', 'Прекрасно! Как у вас?', 'Занят беседой с вами!', 'Чувствую себя отлично!'],
        'default': ['Извините, я не понял ваш вопрос.', 'Можете переформулировать?', 'К сожалению, не могу ответить на это.']
    }

    if intent in responses:
        answer = random.choice(responses[intent])
    else:
        answer = random.choice(responses['default'])
    
    return answer


print('Бот готов к работе')

while True:
    question = input('Вы сказали: ')
    if question == 'стоп':
        break
    answer = bot(question)
    print(answer)
```
