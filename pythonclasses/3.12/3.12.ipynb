```python
import PyPDF2
```

```python
pdf_writer = PyPDF2.PdfWriter()

for i in range(1, 4):
    pdf_file = open(f'file{i}.pdf', 'rb')
    pdf_reader = PyPDF2.PdfReader(pdf_file)
    # print(len(pdf_reader.pages))
    # text1 = pdf_reader.pages[0].extract_text()
    # print(text1)
    for pdf_page in pdf_reader.pages[1:]:
        # text2 = pdf_page.extract_text()
        # print(text2.strip())
        pdf_writer.add_page(pdf_page)
    pdf_file.close()
# merge = PyPDF2.PdfMerger(pdf_file)
# print(merge)

#Записываем writer в файл
with open('Beautiful_story.pdf', 'wb') as f:
    pdf_writer.write(f)



```

```python
pdf_file = open('text.pdf', 'rb')
pdf_reader = PyPDF2.PdfReader(pdf_file)
pdf_page = pdf_reader.pages[0]

watermark_file = open('watermark.pdf', 'rb')
watermark_reader = PyPDF2.PdfReader(watermark_file)
watermark_page = watermark_reader.pages[0]

watermark_page.merge_page(pdf_page)
pdf_writer = PyPDF2.PdfWriter()
pdf_writer.add_page(watermark_page)

with open('watermark_secret.pdf', 'wb') as f:
    pdf_writer.write(f)
```
