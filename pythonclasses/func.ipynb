```python
def calculator(a, operation, b):
    result = 'Что-то пошло не так'
    if operation == '+':
        result = a+b
    elif operation == '-':
        result = a-b
    elif operation == '/':
        result = a/b
    if operation == '*':
        result = a*b
    full_answer = f'{a}{operation}{b} = {result}'
    return full_answer
while True:
    a = int(input('num1'))
    operation = input('sign, exit sign = !')
    b = int(input('num2'))
    res1 = calculator(a, operation, b)
    print(res1)
    if operation == '!':
        print('Goodbye!')
        break
```
