```python
class Pet:
    '''Домашнее животное'''
    legs = 4
    has_tail = True

    def __init__(self, name, awards=0):
        self.name = name
        self.awards = awards

    def __str__(self):
        return (f'{self.name}, у которого(ой) количество ног: {self.legs}, '
                f'есть хвост: {self.has_tail}, медалей: {self.awards}')

class Cat(Pet):
    '''Кошка - является домашним животным'''

    def sound(self):
        print('Мяу!')

class Bobcat(Cat):
    '''Рысь - не является домашним животным'''
    has_tail = False

    def sound(self):
        super().sound()
        print('Страшное Мяу!')

cat1 = Cat('Виталик', 50)
cat1.sound()
print(cat1)

bobcat1 = Bobcat('Злой', 1)
bobcat1.sound()
print(bobcat1)

bobcat2 = Bobcat('Зайка', 3)
bobcat2.sound()
print('У Рыси появился хвост!')
bobcat2.has_tail = True
print(bobcat2)
```

```python
class House:
    def __init__(self):
        self.money = 100
        self.food = 50
        self.dirt = 0

    def __str__(self):
        return f"Дом: денег {self.money}, еды {self.food}, грязи {self.dirt}"

class Human:
    def __init__(self, name):
        self.name = name
        self.satiety = 30
        self.happiness = 100

    def eat(self, house):
        food_to_eat = 30 if house.food >= 30 else house.food
        house.food -= food_to_eat
        self.satiety += food_to_eat
        if self.satiety > 100:
            self.satiety = 100

    def update_happiness(self, house):
        if house.dirt > 90:
            self.happiness -= 20
        if house.food < 10:
            self.happiness -= 15
        if house.money < 50:
            self.happiness -= 10
        if self.happiness < 10:
            self.happiness = 10
            print(f"{self.name} не может больше радоваться и умирает от депрессии!")

    def __str__(self):
        return f"{self.name}: сытость {self.satiety}, счастье {self.happiness}"

class Wife(Human):
    def __init__(self, name):
        super().__init__(name)
        self.coats_bought = 0

    def act(self, house):
        if self.satiety <= 10:
            self.eat(house)
        elif house.food < 30:
            if house.money >= 10:
                food_to_buy = house.money // 10 * 10
                house.food += food_to_buy
                house.money -= food_to_buy
                self.satiety -= 10
            else:
                print(f"{self.name} не смогла купить продукты! Денег нет.")
        elif house.dirt > 90:
            house.dirt = house.dirt - 100 if house.dirt - 100 >= 0 else 0
            self.satiety -= 10
        elif self.happiness <= 30:
            if house.money >= 150:
                house.money -= 150
                self.happiness += 60
                self.coats_bought += 1
                self.satiety -= 10
            else:
                print(f"{self.name} не смогла купить шубу! Денег нет.")
        
        self.update_happiness(house)

class Husband(Human):
    def __init__(self, name):
        super().__init__(name)

    def act(self, house):
        if self.satiety <= 10:
            self.eat(house)
        elif house.money < 50:
            house.money += 111
            self.satiety -= 10
        elif self.happiness <= 30:
            self.happiness += 20
            self.satiety -= 10
        else:
            house.money += 111
            self.satiety -= 10

        self.update_happiness(house)

home = House()
serge = Husband(name="Сережа")
masha = Wife(name="Маша")

for day in range(1, 366):
    print(f"================== День {day} ==================")
    home.dirt += 25
    if home.dirt > 50:
        serge.happiness -= 10  
        masha.happiness -= 10 

    serge.act(home)
    masha.act(home)

    print(serge)
    print(masha)
    print(home)

    if serge.satiety <= 0 or serge.happiness < 10:
        print(f"Сережа умер...")
        break
    if masha.satiety <= 0 or masha.happiness < 10:
        print(f"Маша умерла...")
        break
else:
    print("Год прожит успешно!")
    print(f"Заработано денег: {home.money}")
    print(f"Съедено еды: {home.food}")
    print(f"Куплено шуб: {masha.coats_bought}")
```
