```python
import pprint

results = {'Иванова': {'Макулатура': 3, 'Металлолом': 1, 'Морковь': 4},
           'Петров': {'Макулатура': 10, 'Металлолом': 12, 'Морковь': 2},
           'Сидоров': {'Макулатура': 1, 'Металлолом': 1, 'Морковь': 3},
           'Алексеева': {'Макулатура': 1, 'Металлолом': 2, 'Морковь': 8},
           'Джонсон': {'Макулатура': 23, 'Металлолом': 5, 'Морковь': 2}}

f = open('NewResults3.txt')

for line in f:
    if '-' not in line:
        collect = line.strip()
    else:
        items = line.split()
        person = items[0]
        weight = int(items[2])
    if person not in results:
        results[person] = {}
    if collect not in results[person]:
        results[person][collect] = 0
        results[person][collect] += weight

total_carrots = 0
for person in results:
    if 'Морковь' in results[person]:
        total_carrots += results[person]['Морковь']
print("Общее количество собранной моркови:", total_carrots)

max_paper = 0
max_person = ""
for person in results:
    if 'Макулатура' in results[person] and results[person]['Макулатура'] > max_paper:
        max_paper = results[person]['Макулатура']
        max_person = person
print("Больше всего макулатуры собрал:", max_person, "-", max_paper, "кг")

total_boxes = 0
for person in results:
    if 'Ящики' in results[person]:
        total_boxes += results[person]['Ящики']
print("Общее количество собранных ящиков:", total_boxes)

total_weird_items = 0
for person in results:
    if 'Странные предметы' in results[person]:
        total_weird_items += results[person]['Странные предметы']
print("Общее количество собранных странных предметов:", total_weird_items)

total_scrap_metal = 0
for person in results:
    if 'Металлолом' in results[person]:
        total_scrap_metal += results[person]['Металлолом']
print("Общее количество собранного металлолома:", total_scrap_metal)

f.close()

pprint.pprint(results)



```
