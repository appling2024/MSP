```python
import wikipedia

wikipedia.set_lang("ru")

def get_wikipedia_answer(question):
    try:
        answer = wikipedia.summary(question, sentences=3)
        return answer
    except wikipedia.exceptions.DisambiguationError as e:
        return f"Найдено несколько вариантов: {e.options}"
    except wikipedia.exceptions.RedirectError:
        return "Страница была перенаправлена."
    except wikipedia.exceptions.PageError:
        return "Не знаю, что это."
    except Exception as e:
        return f"Произошла ошибка: {str(e)}"

question_1 = "Что такое Сегежа?"
question_2 = "Кто такая Алла Пугачева?"
question_3 = "Кто такой Эпиктет?"

print(get_wikipedia_answer(question_1))
print(get_wikipedia_answer(question_2))
print(get_wikipedia_answer(question_3))
```
