```python
import xml.etree.ElementTree as ET
```

```python
text = []
sent = []
def context(name):
  tree = ET.parse(f'{name}.man.contexts.txt')
  root = tree.getroot()

  #список, содержащий текст
  x = 0
  tok = False
  full_stop = False
  for elem in root.iter():
     x += 1
     if x > 5:
      text.append(elem.tail)
  #print(text)

  #разбиваем текст на предложения
  y = 0
  j = 0
  sent1 = []
  for i in text:
    if y < 3:
      sent1.append(i)
      if (i == '.' or i == '?') or full_stop == True:
        full_stop = True
        y += 1
    else:
      y = 0
      full_stop = False
      #sent += (sent1,)
      sent.append(sent1)
      j += 1
      sent1 = []

  #вычищаем лишнее, двигаем точки
  for sent1 in sent:
    moved = 0
    for i,word in enumerate(sent1):
      if word == None or word == '\n' or word == '\n\n' or word == ' \t' or word == '\t ' or word == '\t :':
        sent1[i] = ' '
      if word == '\t.':
        sent1[i] = '.'
      if word == '.' or word == '?':
        sent1[i],sent1[-1] = ' ',word

  return sent
```

```python
def search_token(name,token):
  tree = ET.parse(f'{name}.man.contexts.txt')
  root = tree.getroot()

  context(name)

  #cчитаем количество токенов во всём тексте
  token_number = 0
  for elem in root.iter():
    if elem.tail == token:
      token_number += 1

  print(f'{token}: всего найдено {token_number}')

  num = 0
  #выводим контекст
  if token_number > 1:
      sent_context = ''
      for sent1 in sent:
        if num != token_number:
          sent_context = ''
          if token in sent1:
            for i in sent1:
              sent_context += i
            print(f'{num+1})', sent_context)
            num += 1
  else:
    sent_context = ''
    for sent1 in sent:
        if token in sent1:
          for i in sent1:
            sent_context += i
          print('1) ', sent_context)
          break
```

```python
def search_lemma(name,lemma):
  tree = ET.parse(f'{name}.man.contexts.txt')
  root = tree.getroot()

  context(name)

  #cчитаем количество лемм во всём тексте
  lemma_number = 0
  tokens = []
  for elem in root.iter():
    if lemma == elem.attrib.get('lex'):
      lemma_number += 1
      tokens.append(elem.tail)
  print(tokens)

  print(f'{lemma}: всего найдено {lemma_number}')

  #выводим контекст
  num = 1
  for token in tokens:
    if len(tokens) > 1:
      sent_context = ''
      for sent1 in sent:
        if num <= len(tokens):
          sent_context = ''
          if token in sent1:
            for i in sent1:
              sent_context += i
            print(f'{num})', sent_context)
            num += 1
        else:
          break
    else:
      sent_context = ''
      for sent1 in sent:
        if token in sent1:
          for i in sent1:
            sent_context += i
          print('1) ', sent_context)
          break
```

```python
def search_tag(name,tag):
  tree = ET.parse(f'{name}.man.contexts.txt')
  root = tree.getroot()

  context(name)

  #считаем встречаемость тэга в тексте
  tag_number = 0
  tokens = []
  for elem in root.iter():
    if tag in str(elem.attrib.get('gr')) or tag in str(elem.attrib.get('sem')):
      tag_number += 1
      tokens.append(elem.tail)
  print(tokens)

  print(f'{tag}: всего найдено {tag_number}')

  #выводим контекст
  num = 1
  for token in tokens:
    if len(tokens) > 1:
      sent_context = ''
      for sent1 in sent:
        if num <= len(tokens):
          sent_context = ''
          if token in sent1:
            for i in sent1:
              sent_context += i
            print(f'{num})', sent_context)
            num += 1
        else:
          break
    else:
      sent_context = ''
      for sent1 in sent:
        if token in sent1:
          for i in sent1:
            sent_context += i
          print('1) ', sent_context)
          break
```
