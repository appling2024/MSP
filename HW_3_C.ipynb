```python
import nltk
from nltk import CFG, ChartParser
```

```python
grammar = CFG.fromstring("""
    S -> NP VP
    VP -> Neg Verb | Neg Verb NP | Verb Neg | Verb NP Neg
    NP -> Det N | Neg N | N
    Det -> "этот" | "тот"
    N -> "человек" | "ничего" | "никто" | "машина"
    Verb -> "идет" | "говорит" | "делает" | "есть"
    Neg -> "не" | "ни" | "нет" | "никто" | "ничего"
""")
```

```python
parser = ChartParser(grammar)
```

```python
sentences = [
    "никто ничего не делает",
    "человек не идет",
    "ничего нет",
    "этот человек не говорит",
    "никто не делает ничего"
]
```

```python
for sentence in sentences:
    tokens = sentence.split()
    print(f"\nСинтаксический анализ предложения: {' '.join(tokens)}")
    try:
        for tree in parser.parse(tokens):
            print(tree)
    except ValueError:
        print("Не удалось анализировать предложение.")
```
