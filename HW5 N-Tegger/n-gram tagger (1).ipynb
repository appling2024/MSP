```python
import nltk
from nltk.corpus import brown
nltk.download('brown')
```

```python
brown_tagged_sents = brown.tagged_sents(categories='news')
brown_sents = brown.sents(categories='news')
```

```python
brown_tagged_sents = brown.tagged_sents(categories='news')
print(brown_tagged_sents)
```

```python
size = int(len(brown_tagged_sents) * 0.9)
train_sents = brown_tagged_sents[:size]
test_sents = brown_tagged_sents[size:]
```

```python
unigram_tagger = nltk.UnigramTagger(train_sents)
bigram_tagger = nltk.BigramTagger(train_sents)
trigram_tagger = nltk.TrigramTagger(train_sents)
```

```python
unigram_tagger.tag(brown_sents[125])
```

```python
bigram_tagger.tag(brown_sents[125])
```

```python
trigram_tagger.tag(brown_sents[125])
```

```python
print('Unigram:\t', unigram_tagger.accuracy(test_sents),
      '\nBigram:\t\t', bigram_tagger.accuracy(test_sents),
      '\nTrigram:\t', trigram_tagger.accuracy(test_sents))
```
