# Для русского

```python
from nltk.tag import BigramTagger, UnigramTagger, TrigramTagger
from nltk.tag.brill_trainer import BrillTaggerTrainer
from nltk.tag.brill import Template, Pos, Word
```

```python
!wget https://raw.githubusercontent.com/appling2024/data/refs/heads/main/GSD_train.txt
!wget https://raw.githubusercontent.com/appling2024/data/refs/heads/main/GSD_test.txt
```

```python
tagged_sents, tagged_sent = [], []

with open("GSD_train.txt", "r", encoding="utf-8") as f:
    for line in f:
        line = line.split("\t")
        if len(line) > 1:
            tagged_sent.append((line[1], line[3]))
        else:
            tagged_sents.append(tagged_sent)
            tagged_sent = []
            continue

print(tagged_sents)
```

```python
tagged_sents = tagged_sents[:]
unigram_tagger = UnigramTagger(tagged_sents)
bigram_tagger = BigramTagger(tagged_sents, backoff=unigram_tagger)
trigram_tagger = TrigramTagger(tagged_sents, backoff=bigram_tagger)
```

```python
Template._cleartemplates()
templates = fntbl37()
tagger_ru = bt.BrillTaggerTrainer(trigram_tagger, templates, trace=3)
tagger_ru = tagger_ru.train(tagged_sents, max_rules= 1100)
```

# Тестирование на своих фразах

```python
print(tagger_ru.tag(nltk.word_tokenize("Снег в этом году начался слишком ранно и все тропинки покрылись льдом")))
```

# Тестирование на тестовом GSD

```python
correct, incorrect = 0, 0
incorrect_tokens = Counter()

with open("GSD_test.txt", "r", encoding="utf-8") as f:
    for line in f:
        line = line.split("\t")
        if len(line) > 1:
            token = line[1]
            gold_tag = line[3]
            model_tag = tagger_ru.tag([token])
            if gold_tag == model_tag[0][1]:
                correct += 1
                print(gold_tag, model_tag[0][1])
            else:
                incorrect += 1
                incorrect_tokens[gold_tag] = model_tag[0][1]

all_tokens = correct + incorrect
correct_ratio = correct / all_tokens
print("Общее число токенов - ", all_tokens)
print("Доля правильно размеченных токенов от общего числа - ", correct_ratio)
print('Неправильно определил: ', incorrect_tokens)
```
