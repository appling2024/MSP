```python
import os
import re
import pandas as pd
import yake  

def load_text(filepath):
    with open(filepath, 'r', encoding='utf-8') as f:
        return f.read()

def extract_keywords(text, top_n=10):
    kw_extractor = yake.KeywordExtractor(lan="ru", n=3, dedupLim=0.9, top=top_n, features=None)
    keywords = kw_extractor.extract_keywords(text)
    return [kw[0] for kw in keywords]

def generate_kw_table(root_dir, output_excel_path):
    pattern = re.compile(r'^(\w+)_IMS_(20\d{2})_rus\.txt$')
    records = []

    for dirpath, _, filenames in os.walk(root_dir):
        for file in filenames:
            match = pattern.match(file)
            if not match:
                continue  

            author, year = match.group(1), match.group(2)
            full_path = os.path.join(dirpath, file)
            try:
                text = load_text(full_path)
                keywords = extract_keywords(text)
                records.append({
                    'Файл': file,
                    'Автор': author,
                    'Год': year,
                    'НКВ': '; '.join(keywords)
                })
                print(f"Обработано: {file}")
            except Exception as e:
                print(f"Ошибка при обработке {file}: {e}")

    df = pd.DataFrame(records)
    df.to_excel(output_excel_path, index=False)
    print(f"\nСохранено: {output_excel_path}")

if __name__ == '__main__':
    input_folder = 'IMS2013-2024'
    output_file = 'Generated_KW.xlsx'
    generate_kw_table(input_folder, output_file)
```
