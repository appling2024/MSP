### Тут я тренируюсь на одном предложении-контексте

Все контексты состоят из слов, которые заключены в теги \<w>\</w>, а вся информация (лемма, грамматика и семантика) заключена в теге \<ana> (атрибуты lex, gr, sem соответственно).

Хотим доставать все штуки вида "\<w>\<ana lex='плеск' gr='S,m,inan,norm=(nom,sg|acc,sg)' sem='r:abstr t:sound '/>плеск\</w>".

```python
import re
pattern = r'<w>.+?</w>' # с помощью ? мы можем сделать наше регулярное выражение нежадным!!
words = re.findall(pattern, context)
for word in words:
  print(word)
```

```python
tokens = []
for word in words:

  lex_pattern = r'lex=\'.+?\''
  lex_raw = re.search(lex_pattern, word).group()
  lex = lex_raw.split('=')[1].strip("'")  # нашли лексему
  print('Лексема: ' + lex)

  form_pattern = '>[А-Яа-я]+<'
  form = re.search(form_pattern, word).group().strip('<>')  # нашли словоформу
  tokens.append(form) # добавим найденную словоформу в список всех словоформ данного предложения
  print('Словоформа: ' + form)

  sem_pattern = r'sem=\'.+?\''
  sem_raw = re.search(sem_pattern, word)
  if sem_raw:
    sem = sem_raw.group().split('=')[1].strip("'")  # нашли семантику
    print('Семантика: ' + sem)



  print('_____________________')

# Соберём все слова в одно предложение-контекст
sentence = ' '.join(tokens)
print(sentence)
```

### Предобработака всех контекстов

```python
""""Функция, которая на вход получает контекст в сыром виде (то есть со всеми тегами и ненужными скобочками) и преобразует его так, чтобы по нему было удобно искать"""
import re
def preprocess(context):
  info = {'lex': [], 'sem': [], 'tokens': [], 'sentence': ''} # словарь вида {lex: список лемм, sem: список семантических тегов, tokens: список словоформ, sentence: контекст в виде удобочитаемой строки}
  pattern = r'<w>.+?</w>' # с помощью ? мы можем сделать наше регулярное выражение нежадным!!
  words = re.findall(pattern, context) # добыли "слова", но тоже пока что в некрасивом виде

  tokens = []
  for word in words:
    lex_pattern = r'lex=\'.+?\''
    lex_raw = re.search(lex_pattern, word)
    if lex_raw:
      lex = lex_raw.group().split('=')[1].strip("'")  # нашли лексему
      info['lex'].append(lex) # добавили лексему в наш словарь


    sem_pattern = r'sem=\'.+?\''
    sem_raw = re.search(sem_pattern, word)
    if sem_raw:  # проверяем, что совпадение найдено (не у всех слов есть семантика)
      sem = sem_raw.group().split('=')[1].strip("'").split()  # нашли семантику, причём в список добавляем семантический тег не единой строкой, а списком, чтобы иметь возможность искать по отдельным тегам, а не по всему в куче
      info["sem"].append(sem) # добавили семантические теги в наш словарь

    form_pattern = '>\w+<'
    form_raw = re.search(form_pattern, word)
    if form_raw:
      form = form_raw.group().strip('<>')  # нашли словоформу
      info['tokens'].append(form) # добавили словоформу в наш словарь
      tokens.append(form) # добавили найденную словоформу в список всех словоформ данного предложения

  # Соберём все слова в одно предложение-контекст
  sentence = ' '.join(tokens)
  info['sentence'] = sentence # добавили полученное предложение в наш словарь

  return info # возвращаем словарь с полной информации о данном контексте
```

Сформируем словарь со всеми контекстами из файла.

```python
from tqdm import tqdm
full_info = {} # словарь вида индекс контекста: словарь с информацией для этого контекста
for i in tqdm(range(len(metla))):
  context = metla[i]
  full_info[i] = preprocess(context)
```

### Поиск по токену

Функция, которая на вход получает токен (словоформу), а на выход отдаёт контексты, в которых встречается эта словоформа.

```python
def find_by_token(token):
  good_contexts = [] # список предложений, содержащий данную словоформу
  for i in range(len(full_info)):
    tokens = full_info[i]['tokens']
    sentence = full_info[i]['sentence']
    if token in tokens:
      good_contexts.append(sentence)

  return good_contexts
```

```python
find_by_token('метла')
```

### Поиск по лемме

Подаём на вход лемму, на выходе получаем предложения, содержащие любые формы данной леммы.

```python
def find_by_lemma(lemma):
  good_contexts = []
  for i in range(len(full_info)):
    lemmas = full_info[i]['lex']
    sentence = full_info[i]['sentence']
    if lemma in lemmas:
      good_contexts.append(sentence)

  return good_contexts
```

```python
find_by_lemma('метла')
```

### Поиск по отдельному семантическому тегу

```python
def find_by_tag(tag):
  good_contexts = []
  for i in range(len(full_info)):
    sem = full_info[i]['sem'] # список со списками тегов
    sentence = full_info[i]['sentence']
    for tag_set in sem:
      if tag in tag_set:
        good_contexts.append(sentence)
        continue

  return list(set(good_contexts))
```

```python
find_by_tag('t:appearance')
```

### Поиск по набору семантических тегов

Хотим вывести предложения, в которых встречаются слова, обладающие именно заданной комбинацией семантических тегов (не больше и не меньше).

Вводить надо несколько тегов через пробел.

```python
def find_by_tags(tags):
  good_contexts = []
  tagset = tags.split() # получаем список со всеми тегами заданного тегсета
  for i in range(len(full_info)):
    sem = full_info[i]['sem']
    sentence = full_info[i]['sentence']
    for t in sem: # перебираем наборы тегов для каждого слова текущего предложения
      if set(tagset) == set(t):
        good_contexts.append(sentence)

  return good_contexts
```

## Предобработка данных

**Файл vostorg.man.contexts.txt**

```python
import pandas as pd
from bs4 import BeautifulSoup as bs
```

```python
with open('vostorg.man.contexts.txt', encoding='ANSI') as f:
  soup = bs(f, 'lxml')
```

Файл состоит из параграфов (абзацев?), заключённых между тегами \<p>\</p>. В каждом параграфе - несколько предложений, заключённых между тегами \<se>\</se>.

```python
# Сколько параграфов? Вычленим их и посчитаем
paragraphs = soup.find_all('p')
print(f'Количество параграфов: {len(paragraphs)}')
```

```python
# Сколько предложений? Тоже вытащим их из документа и посчитаем
sentences = soup.find_all('se') # "сырые" предложения (не только тексты предложений, но и всякие теги и пр.)
sents = [] # тут уже читабельные предложения (видимо, с помощью .text мы достаём всё, что находится между тегами)
for sentence in sentences:
  sents.append(sentence.text)
print(f'Количество предложений: {len(sents)}')
```

```python
repeated_sents = []  # чтобы каждому слову документа поставить в соответствие предложение-контекст, надо сохранить столько предложений, сколько у нас слов в документе
tokens = []
lemmas = []
sems = []
grams = []
index = [] # индекс слова в предложении (понадобится в дальнейшем при вычленении узких контекстов)
for sentence in sentences:
  words = sentence.find_all('w') # проходим по всем предложениям документа и вычленяем оттуда слова
  for i in range(len(words)):
    word = words[i]
    index.append(i)
    tokens.append(word.text.lower()) # текст тега <w> - сам токен
    ana = word.find_next('ana') # первый тег <ana> после текущего тега
    # теги <ana> имеют вид <ana lex="образование" sem="der:v r:abstr der:s0 t:be:appear" gr="S n inan sg gen"/>образования</w>
    # можем достать значение того или иного атрибута, обращаясь к нему по его имени (как по ключу в словаре)
    lemmas.append(ana['lex'])
    if 'sem' in ana.attrs: # ana.attrs - список атрибутов тега ana (не у всех слов есть семантика, например у служебных слов её нет)
      sems.append(ana['sem'])
    else:
      sems.append(None)
    grams.append(ana['gr'])
    repeated_sents.append(sentence.text)
```

```python
# проверим, что у нас получились списки одинаковой длины
len(repeated_sents) == len(tokens) == len(lemmas) == len(sems) == len(grams) == len(index)
```

Упакуем всю информацию в таблицу pandas.

```python
df = pd.DataFrame(zip(tokens, lemmas, grams, sems, repeated_sents, index), columns=['Tokens', 'Lemmas', 'Grammar tags', 'Semantic tags', 'Sentences', 'Index'])
df
```

```python
print(f'First token: {df["Tokens"][0]}')
if df['Tokens'][0] == 'понимая':  
    print(f'Lemma: {df["Lemmas"][0]}')
    print(f'Semantics: {df["Semantic tags"][0]}')
    print(f'Sentence: {df["Sentences"][0]}')
```

### Контекст - не всё предложение

```python
def find_token_in_context(token, size): # token - искомый токен, size - размер контекстного окна справа и слева от целевого слова
# если упираемся в начало и/или конец предложений - штош, ничего не поделаешь
  contexts = []
  for i in range(df.shape[0]):
    if df['Tokens'][i] == token:
      sentence = df['Sentences'][0].split() # теперь предложение - это список слов
      ind = df['Index'][i] # индекс слова в его предложении
      previous = len(sentence[:ind]) # сколько слов в предложении предшествуют целевому слову
      next = len(sentence[ind+1:]) # сколько слов в предложении следуют за целевым словом
      previous_context = [] # искомый предшествующий контекст
      next_context = [] # искомый последующий контекст
      # соберём предшествующий контекст
      if previous >= size:  # если перед целевым словом стоит слов не меньше, чем размер контекстного окна
        for j in range(ind - size, ind):
          previous_context.append(sentence[j])
      elif 0 < previous < size:
        for j in range(ind - previous, ind):
          previous_context.append(sentence[j])
      # соберём последующий контекст
      if next >= size:
        for j in range(ind + 1, ind + size + 1):
          next_context.append(sentence[j])
      elif 0 < next < size:
        for j in range(ind + 1, ind + 1 + next):
          next_context.append(sentence[i])

      # теперь надо собрать всё это в единый кусочек: предшествующий контекст + целевое слово + последующий контекст
      pr_cont = ' '.join(previous_context)
      n_cont = ' '.join(next_context)
      full_cont = ' '.join([pr_cont, token, n_cont])
      contexts.append(full_cont)

  return contexts
```
