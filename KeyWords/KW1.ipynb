TermExtractor

```python
!pip install rutermextract
```

```python
from rutermextract import TermExtractor
```

```python
term_extractor = TermExtractor()
```

```python
with open('ChizhikEgorov_IMS_2022.txt', 'r', encoding='utf-8') as f:
    text = f.read()
```

```python
for term in term_extractor(text):
    print(term.normalized, term.count)
```

```python
with open('RuTermExtract_keywords.csv', 'w', encoding='utf-8') as f2:
    for term in term_extractor(text):
        f2.write(str(term.normalized)+";"+str(term.count)+"\n")
```

```python
from keybert import KeyBERT
```

```python
with open('ChizhikEgorov_IMS_2022.txt', 'r', encoding='utf-8') as f:
    doc = f.read()
```

```python
kw_model = KeyBERT()
```

```python
keywords = kw_model.extract_keywords(doc)
```

```python
keywords
```

```python
keywords = kw_model.extract_keywords(doc, top_n=10)
keywords
```

```python
with open('KeyBERT_top_10.csv', 'w', encoding='utf-8') as f2:
    for eachkwrd in keywords:
        f2.write(str(eachkwrd[0])+";"+str(eachkwrd[1])+"\n")
```

```python
keywords = kw_model.extract_keywords(doc, keyphrase_ngram_range=(1, 2), top_n=10)
keywords
```

```python
with open('KeyBERT_top_10_bigram.csv', 'w', encoding='utf-8') as f2:
    for eachkwrd in keywords:
        f2.write(str(eachkwrd[0])+";"+str(eachkwrd[1])+"\n")
```

```python
with open('KeyBERT_top_10_trigram.csv', 'w', encoding='utf-8') as f2:
    for eachkwrd in keywords:
        f2.write(str(eachkwrd[0])+";"+str(eachkwrd[1])+"\n")
```
