```python
from nltk.stem import RegexpStemmer

regexp = RegexpStemmer('minen$|t$|ssa$|lla$|ksi$|ta$|ri$|ja$|^epä', min=4)

words = [
    'kirjoittaminen',  # написание
    'kirjoitukset',    # записи
    'juoksija',        # бегун
    'Turussa',     # в Турку
    'asemalla',         # на станции
    'koirat',      # собаки
    'kaloja',          # рыбы
    'epäselvä',       # непонятный
    'telkkari',      # телек
    'puutarhassa'      # в саду
]

for word in words:
    print(word, "--->", regexp.stem(word))
```

```python
from nltk.stem.snowball import SnowballStemmer

stemmer = SnowballStemmer("finnish")

words = ['kissojen', 'koirien', 'talossa', 'metsästä', 'kirjoja', 'opiskelijoiden', 'juoksussa']

for word in words:
    print(word, "--->", stemmer.stem(word))
```

```python
from nltk.stem import SnowballStemmer, RegexpStemmer

snowball = SnowballStemmer(language='finnish')  # Snowball стеммер для финского
regexp = RegexpStemmer(r'(minen$|t$|ssa$|lla$|ksi$|ta$|ri$|ja$|^epä)', min=4)  # Регулярное выражение для финского

word_list = [
    'kirjoittaminen',  # написание
    'kirjoitukset',    # записи
    'juoksija',        # бегун
    'Turussa',         # в Турку
    'asemalla',        # на станции
    'koirat',          # собаки
    'kaloja',          # рыбы
    'epäselvä',        # непонятный
    'telkkari',        # телек
    'puutarhassa'      # в саду
]

print("{0:20}{1:20}{2:30}".format("Word", "Snowball Stemmer", "Regexp Stemmer"))

for word in word_list:
    print("{0:20}{1:20}{2:30}".format(word, snowball.stem(word), regexp.stem(word)))
```
