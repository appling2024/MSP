```python
import yake
from keybert import KeyBERT
from rutermextract import TermExtractor

file_main = "C:/Users/user/Documents/GitHub/MSP/keywords extractor/Belkin_IMS_2024_rus.txt"
file_abstract = "C:/Users/user/Documents/GitHub/MSP/keywords extractor/Belkin_IMS_2024_Abstract_rus.txt"
file_output = "C:/Users/user/Documents/GitHub/MSP/keywords extractor/result.txt"

with open(file_main, "r", encoding="utf-8") as file:
    text_main = file.read()

with open(file_abstract, "r", encoding="utf-8") as file:
    text_abstract = file.read()

def extract_keywords(text):
    kw_extractor = yake.KeywordExtractor(lan="ru", n=3, top=10)
    yake_keywords = [kw[0] for kw in kw_extractor.extract_keywords(text)]

    kw_model = KeyBERT()
    keybert_keywords = [kw[0] for kw in kw_model.extract_keywords(text, top_n=10)]

    term_extractor = TermExtractor()
    ruterm_keywords = [term.normalized for term in term_extractor(text)[:10]]

    return yake_keywords, keybert_keywords, ruterm_keywords

yake_main, keybert_main, ruterm_main = extract_keywords(text_main)
yake_abs, keybert_abs, ruterm_abs = extract_keywords(text_abstract)

with open(file_output, "w", encoding="utf-8") as out_file:
    out_file.write("=== Основной текст ===\n")
    out_file.write("YAKE:\n" + ", ".join(yake_main) + "\n\n")
    out_file.write("KeyBERT:\n" + ", ".join(keybert_main) + "\n\n")
    out_file.write("RuTermExtract:\n" + ", ".join(ruterm_main) + "\n\n")

    out_file.write("=== Аннотация ===\n")
    out_file.write("YAKE:\n" + ", ".join(yake_abs) + "\n\n")
    out_file.write("KeyBERT:\n" + ", ".join(keybert_abs) + "\n\n")
    out_file.write("RuTermExtract:\n" + ", ".join(ruterm_abs) + "\n")

print("Ключевые слова успешно извлечены и сохранены.")
```
