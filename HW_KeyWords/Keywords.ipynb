```python
# Определение путей к файлам
file_main = "C:\\Users\\79040\\OneDrive\\Рабочий стол\\MSP\\HW_KeyWords\\ChizhikEgorov_IMS_2022.txt"
file_abstract = "C:\\Users\\79040\\OneDrive\\Рабочий стол\\MSP\\HW_KeyWords\\ChizhikEgorov_IMS_2022_Abstract.txt"
file_keywords = "c:\\Users\\79040\\OneDrive\\Рабочий стол\\MSP\\HW_KeyWords\\ChizhikEgorov_IMS_2022_KW_rus.txt"
file_output = "C:\\Users\\79040\\OneDrive\\Рабочий стол\\MSP\\HW_KeyWords\\results.txt"
```

```python
# Чтение основного текста
with open(file_main, "r", encoding="utf-8") as file:
    text_main = file.read()
```

```python
# Чтение аннотации
with open(file_abstract, "r", encoding="utf-8") as file:
    text_abstract = file.read()
```

```python
# Чтение эталонных ключевых слов и их обработка
with open(file_keywords, "r", encoding="utf-8") as file:
    reference_keywords_raw = file.read().strip()
    reference_keywords = set(reference_keywords_raw.lower().replace(";", " ").split())
```

```python
def extract_keywords(text):
    """
    Извлекает ключевые слова из текста с помощью трех различных методов:
    - YAKE
    - KeyBERT
    - RuTermExtract
    """
    # Извлечение ключевых слов с помощью YAKE
    kw_extractor = yake.KeywordExtractor(lan="ru", n=3, top=10)
    yake_keywords = set(word for kw in kw_extractor.extract_keywords(text) for word in kw[0].lower().split())

    # Извлечение ключевых слов с помощью KeyBERT
    kw_model = KeyBERT()
    keybert_keywords = set(word for kw in kw_model.extract_keywords(text, top_n=10) for word in kw[0].lower().split())

    # Извлечение ключевых слов с помощью RuTermExtract
    term_extractor = TermExtractor()
    ruterm_keywords = set(word for term in term_extractor(text)[:10] for word in term.normalized.lower().split())

    return yake_keywords, keybert_keywords, ruterm_keywords
```

```python
def calculate(extracted, reference):
    """
    Вычисляет процент совпадения между извлеченными и эталонными ключевыми словами
    """
    matches = extracted & reference
    return round((len(matches) / len(reference) * 100) if reference else 0, 2), matches

# Извлечение ключевых слов из основного текста и аннотации
yake_main, keybert_main, ruterm_main = extract_keywords(text_main)
yake_abs, keybert_abs, ruterm_abs = extract_keywords(text_abstract)

# Расчет совпадений для основного текста
yake_main_match, yake_main_words = calculate(yake_main, reference_keywords)
keybert_main_match, keybert_main_words = calculate(keybert_main, reference_keywords)
ruterm_main_match, ruterm_main_words = calculate(ruterm_main, reference_keywords)

# Расчет совпадений для аннотации
yake_abs_match, yake_abs_words = calculate(yake_abs, reference_keywords)
keybert_abs_match, keybert_abs_words = calculate(keybert_abs, reference_keywords)
ruterm_abs_match, ruterm_abs_words = calculate(ruterm_abs, reference_keywords)
```

```python
# Запись результатов в файл
with open(file_output, "w", encoding="utf-8") as out_file:
    # Запись эталонных ключевых слов
    out_file.write("Эталонные ключевые слова\n")
    out_file.write(reference_keywords_raw + "\n\n")

    # Запись результатов для основного текста
    out_file.write("Основной текст\n")
    out_file.write(f"YAKE ({yake_main_match}% совпадений):\n" + ", ".join(yake_main) + "\n")
    out_file.write(f"Совпавшие слова: {', '.join(yake_main_words)}\n\n")

    out_file.write(f"KeyBERT ({keybert_main_match}% совпадений):\n" + ", ".join(keybert_main) + "\n")
    out_file.write(f"Совпавшие слова: {', '.join(keybert_main_words)}\n\n")

    out_file.write(f"RuTermExtract ({ruterm_main_match}% совпадений):\n" + ", ".join(ruterm_main) + "\n")
    out_file.write(f"Совпавшие слова: {', '.join(ruterm_main_words)}\n\n")

    # Запись результатов для аннотации
    out_file.write("Аннотация\n")
    out_file.write(f"YAKE ({yake_abs_match}% совпадений):\n" + ", ".join(yake_abs) + "\n")
    out_file.write(f"Совпавшие слова: {', '.join(yake_abs_words)}\n\n")

    out_file.write(f"KeyBERT ({keybert_abs_match}% совпадений):\n" + ", ".join(keybert_abs) + "\n")
    out_file.write(f"Совпавшие слова: {', '.join(keybert_abs_words)}\n\n")

    out_file.write(f"RuTermExtract ({ruterm_abs_match}% совпадений):\n" + ", ".join(ruterm_abs) + "\n")
    out_file.write(f"Совпавшие слова: {', '.join(ruterm_abs_words)}\n")

print("вери гуд")
```
