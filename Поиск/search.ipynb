```python
import pymorphy3
from typing import List, Dict

# Инициализация морфологического анализатора
morph = pymorphy3.MorphAnalyzer()

# Корпус в виде списка предложений с размеченными токенами
# Пример структуры: каждый токен содержит лемму, морфологический тэг и специфический тэг
corpus = [
    [{"word": "кот", "lemma": "кот", "morph_tag": "NOUN,masc", "custom_tag": "животное"},
     {"word": "бегал", "lemma": "бегать", "morph_tag": "VERB,past,masc", "custom_tag": "действие"}],
    [{"word": "собака", "lemma": "собака", "morph_tag": "NOUN,fem", "custom_tag": "животное"},
     {"word": "спала", "lemma": "спать", "morph_tag": "VERB,past,fem", "custom_tag": "действие"}],
]

def search_corpus(corpus: List[List[Dict[str, str]]], lemma: str = None, morph_tag: str = None, custom_tag: str = None) -> List[Dict[str, str]]:
    """
    Функция для поиска в корпусе по лемме, морфологическому тэгу и специфическому тэгу.
    :param corpus: Корпус текста в виде списка предложений с токенами.
    :param lemma: Искомая лемма.
    :param morph_tag: Искомый морфологический тэг.
    :param custom_tag: Искомый специфический тэг.
    :return: Список токенов, соответствующих критериям.
    """
    results = []
    for sentence in corpus:
        for token in sentence:
            if (
                (lemma is None or token["lemma"] == lemma) and
                (morph_tag is None or morph_tag in token["morph_tag"]) and
                (custom_tag is None or token["custom_tag"] == custom_tag)
            ):
                results.append(token)
    return results

# Пример использования функции поиска
result = search_corpus(corpus, lemma="кот", morph_tag="NOUN", custom_tag="животное")

# Вывод результата
print("Результаты поиска:", result)
```
