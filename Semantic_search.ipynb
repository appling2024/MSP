Поиск по семантическому тэгу 

```python
import random
```

Поиск по Core элементам → строго связаны с глаголом (например, субъект, объект, адресат)

```python
def corsem_search(file, sem_role): 
    with open(file, 'r', encoding = 'utf-8') as f:
        next(f)
        results = []
        for line in f:
            items = line.split()
            if sem_role == items[6]:
                results.append(line.strip())
    
        if results:
            sample_size = min(10, len(results))
            random_sample = random.sample(results, sample_size) 
            print('ConstrIndex\tPlace\tPhraseGen\tWordDep\tForm\tRole\tRank\tSem\tRea\tConstrExId\tItemExIndex\tKeyLexemes\n' + "\n".join(random_sample))
        else:
            print('совпадений не найдено')
```

```python
file2 = 'framebank_anno_ex_items.txt'
sem_role1 = 'предмет'
result = corsem_search(file2, sem_role1)
```

Поиск по Non-core элементам → обстоятельства (место, время, причина и т. д.).

```python
def ncorsem_search(file, sem_role):
    with open(file, 'r', encoding = 'utf-8') as f:
        next(f)
        results = []
        for line in f:
            items = line.split()
            if sem_role == items[5]:
                results.append(line.strip())
    
        if results:
            sample_size = min(10, len(results))  # Берём 5 строк, но не больше, чем есть
            random_sample = random.sample(results, sample_size)  # Выбираем случайные строки
            print('ConstrIndex\tExIndex\tPlaceC\tPhrase\tForm\tRole\tType\tKeyLexemes\n' + "\n".join(random_sample))
        else:
            print('совпадений не найдено')
```

```python
file = 'framebank_anno_ex_circ.txt'
sem_role0 = 'узуальность'
result = ncorsem_search(file, sem_role0)
```
