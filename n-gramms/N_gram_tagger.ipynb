```python
import nltk
import pandas as pd
import csv
import os
import pathlib
#path = pathlib.Path('C:\\Users\\Dark Lord\\Download')

with open('GSD_test.txt', 'r') as f:
  test_data = f.read()

data = pd.read_csv(f'GSD_test.txt', sep = '	')
data.to_csv(fr'GSD_test.csv',index =None)
data = pd.read_csv(fr'GSD_test.csv')
del data['1']  # удаление лишних столбцов
del data['3']  #
del data['5']  #
data.to_csv(f'GSD_test_.csv', index=False)
```

```python
tuples = []
with open('GSD_test_.csv', 'r', encoding='utf-8') as f:  # преобразование csv в список кортежей
    read_csv = csv.reader(f, delimiter=",")
    for row in read_csv:
        row = tuple(row)
        tuples.append(row)

tuples = tuple(tuples)
```

```python
gold, pair = [], []

for item in tuples:  # список списков кортежей с разделением по точке
    pair.append(item)
    if item[0] == '.':
        gold.append(pair)
        pair = []

if len(pair) > 0:
    gold.append(pair)
```

```python
size = int(len(gold) * 0.8)  # разделение тренировочных данных
print(size)
train_sents = gold[:size]
test_sents = gold[size:]
```

```python
unigram_tagger = nltk.UnigramTagger(train_sents)  # униграмный теггер: обучение
print(unigram_tagger.accuracy(test_sents))
```

```python
bigram_tagger = nltk.BigramTagger(train_sents)   # биграмный теггер: обучение
print(bigram_tagger.accuracy(test_sents))
```

```python
trigram_tagger = nltk.TrigramTagger(train_sents)  # триграмный теггер: обучение
print(trigram_tagger.accuracy(test_sents))
```

```python
import re
pin =[]
with open(r'pin.txt', encoding='utf-8') as n_f:
    pin = n_f.read()
pin = re.findall(r"[\w']+|[.,!?;…]", pin)
print(pin)
```

```python
brusov_tagged_sents_1 = unigram_tagger.tag(pin)
brusov_tagged_sents_2 = bigram_tagger.tag(pin)
brusov_tagged_sents_3 = trigram_tagger.tag(pin)
print('Unigram_tagger',brusov_tagged_sents_1)
print('Bigram_tagger',brusov_tagged_sents_2)
print('Trigram_tagger',brusov_tagged_sents_3)
```
