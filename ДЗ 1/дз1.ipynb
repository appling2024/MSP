```python
import nltk
import re

# Читаем текст из файла
with open("C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/ДЗ 1/text_finnish_history.txt", "r", encoding="utf-8") as file:
    text = file.read()

cleaned_text = text.strip()
cleaned_text = re.sub(r'open\(".*"\)|file\.write\(.*\)|encoding=".*"|utf-8', '', cleaned_text)

# Скачиваем необходимые ресурсы для NLTK
nltk.download("punkt")

# Токенизация
from nltk.tokenize import word_tokenize
tokens_nltk = word_tokenize(cleaned_text)

# Сохранение результата в файл
with open("C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/ДЗ 1/tokens_nltk.txt", "w", encoding="utf-8") as file:
    file.write("\n".join(tokens_nltk))

print("Токенизация с использованием NLTK завершена. Результат сохранен в tokens_nltk_cleaned.txt.")
```

```python
import spacy
import re

# Читаем текст
with open("C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/ДЗ 1/text_finnish_history.txt", "r", encoding="utf-8") as file:
    text = file.read()

cleaned_text = text.strip()
cleaned_text = re.sub(r'open\(".*"\)|file\.write\(.*\)|encoding=".*"|utf-8', '', cleaned_text)

# Загружаем модель SpaCy
nlp = spacy.load("en_core_web_sm")

# Обрабатываем текст
doc = nlp(cleaned_text)

# Извлекаем токены
tokens_spacy = [token.text for token in doc]

# Сохраняем результат в файл
with open("C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/ДЗ 1/tokens_spacy.txt", "w", encoding="utf-8") as file:
    file.write("\n".join(tokens_spacy))

print("Токенизация с использованием SpaCy завершена.")
```

```python
from sacremoses import MosesTokenizer
import re

# Читаем текст из файла
with open("C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/ДЗ 1/text_finnish_history.txt", "r", encoding="utf-8") as file:
    text = file.read()

cleaned_text = text.strip()
cleaned_text = re.sub(r'open\(".*"\)|file\.write\(.*\)|encoding=".*"|utf-8', '', cleaned_text)

# Токенизация с Moses
moses_tokenizer = MosesTokenizer()
tokens_moses = moses_tokenizer.tokenize(cleaned_text)

#Сохранение результата в файл
with open("C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/ДЗ 1/tokens_moses.txt", "w", encoding="utf-8") as file:
    file.write("\n".join(tokens_moses))

print("Токенизация с использованием Moses завершена")
```

```python
# Читаем токены из файлов
with open("tokens_nltk.txt", "r", encoding="utf-8") as file:
    tokens_nltk = set(file.read().splitlines())

with open("tokens_spacy.txt", "r", encoding="utf-8") as file:
    tokens_spacy = set(file.read().splitlines())

with open("tokens_moses.txt", "r", encoding="utf-8") as file:
    tokens_moses = set(file.read().splitlines())

# Сравниваем токены
print("Уникальные токены NLTK:")
print(tokens_nltk - tokens_spacy - tokens_moses)

print("\nУникальные токены SpaCy:")
print(tokens_spacy - tokens_nltk - tokens_moses)

print("\nУникальные токены Moses:")
print(tokens_moses - tokens_nltk - tokens_spacy)

# Совпадения между всеми библиотеками
common_tokens = tokens_nltk & tokens_spacy & tokens_moses
print("\nОбщие токены для всех библиотек:")
print(common_tokens)
```
