The Flesch Reading Ease formula

это метод оценки, который определяет, насколько легко текст воспринимается читателем.

Формула оценивает удобочитаемость текста, присваивая ему числовой балл от 0 до 100. Более высокие баллы (например, 60-70) указывают на более легкий для чтения текст, в то время как низкие баллы (например, 0-30) указывают на более сложный текст. 
Основные компоненты формулы:
Длина предложения: Количество слов в предложении.
Длина слова: Количество слогов в слове. 

```python
textstat.flesch_reading_ease(text) #standart
```

The Flesch-Kincaid Grade Level

это показатель удобочитаемости, который определяет, какой уровень образования требуется читателю для понимания текста

Чем выше оценка, тем сложнее читать

Flesch-Kincaid Grade Level - это формула, разработанная для оценки сложности текста на основе двух факторов: средней длины предложения и среднего количества слогов в слове. 

```python
textstat.flesch_kincaid_grade(text)
```

The Fog Scale (Gunning FOG Formula)

Индекс туманности учитывает два основных фактора: 
1. Средняя длина предложения в словах:
Более длинные предложения, как правило, сложнее для понимания.
2. Количество "сложных" слов:
Сложные слова - это слова, состоящие из трех и более слогов.
Формула расчета (приблизительно):
Индекс = 0.4 * (средняя длина предложения в словах + процент "сложных" слов). 

Чем выше индекс, тем сложнее текст

The Coleman-Liau Index

В отличие от других формул, которые считают слоги, индекс Колмана-Лиау использует количество символов в словах и предложений на 100 слов. 

Используется следующая формула: 5.89 * (количество символов / количество слов) - 0.3 * (количество предложений / количество слов) - 15.8. 

```python
textstat.coleman_liau_index(text)
```

Linsear Write Formula

The standard Linsear Write metric Lw runs on a 100-word sample:[3]

For each "easy word", defined as words with 2 syllables or less, add 1 point.
For each "hard word", defined as words with 3 syllables or more, add 3 points.
Divide the points by the number of sentences in the 100-word sample.
Adjust the provisional result r:
If r > 20, Lw = r / 2.
If r ≤ 20, Lw = r / 2 - 1.
The result is a "grade level" measure, reflecting the estimated years of education needed to read the text fluently.

```python
textstat.linsear_write_formula(text)
```

McAlpine EFLAW Readability Score

McAlpine EFLAW Readability Score — это специализированная метрика оценки удобочитаемости текста, разработанная Rachel McAlpine для анализа технической документации, руководств и инструкций. Она фокусируется на пяти ключевых аспектах текста, объединённых в аббревиатуру EFLAW:

E (English) — Правильность грамматики и синтаксиса.

F (Facts) — Точность и достоверность информации.

L (Logic) — Логичность структуры и последовательность изложения.

A (Accessibility) — Доступность для целевой аудитории (ясность, отсутствие жаргона).

W (Wordiness) — Лаконичность (отсутствие «воды» в тексте).

Оценка: Каждый параметр (E, F, L, A, W) проверяется отдельно, часто вручную или через чек-листы.

Документ с высоким EFLAW-показателем должен быть:

Грамотным (E),

Без ошибок в данных (F),

Структурированным (L),

Понятным для новичков (A),

Кратким (W).

```python
textstat.mcalpine_eflaw(text)
```

```python
textstat.mcalpine_eflaw(text_eng)
```

10–12: Удовлетворительная ясность, но есть местами сложности для неподготовленной аудитории.

<10: Текст слишком сложен или содержит ошибки.

>15: Очень высокое качество по всем параметрам EFLAW.
