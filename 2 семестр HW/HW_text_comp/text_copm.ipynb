```python
import os
import glob
import pandas as pd
import textstat
from ruts import BasicStats, ReadabilityStats, DiversityStats, MorphStats

def read_text_file(file_path):
    """Чтение текста из файла"""
    with open(file_path, 'r', encoding='utf-8') as file:
        return file.read()


def analyze_text_complexity(text):
    """Анализ сложности текста всеми доступными метриками"""
    results = {}
    
    # Базовые статистики текста
    bs = BasicStats(text)
    results.update(bs.get_stats())
    
    # Статистики удобочитаемости
    rs = ReadabilityStats(text)
    results.update(rs.get_stats())
    
    # Статистики разнообразия лексики
    ds = DiversityStats(text)
    results.update(ds.get_stats())
    
    # Морфологические статистики
    ms = MorphStats(text)
    results.update(ms.get_stats())
    
    # Дополнительные метрики удобочитаемости
    results['flesch_reading_ease'] = textstat.flesch_reading_ease(text)
    results['flesch_kincaid_grade'] = textstat.flesch_kincaid_grade(text)
    results['smog_index'] = textstat.smog_index(text)
    results['coleman_liau_index'] = textstat.coleman_liau_index(text)
    results['automated_readability_index'] = textstat.automated_readability_index(text)
    results['dale_chall_readability_score'] = textstat.dale_chall_readability_score(text)
    results['difficult_words'] = textstat.difficult_words(text)
    results['linsear_write_formula'] = textstat.linsear_write_formula(text)
    results['gunning_fog'] = textstat.gunning_fog(text)
    return results

def main():
    base_path = "C:\\Users\\79040\\OneDrive\\Рабочий стол\\MSP\\2 семестр HW\\HW_text_comp\\IMS2013-2024\\IMS2013-2024"
    selected_texts = []

    # Сбор текстов по годам
    for year in range(2013, 2025):
        year_folder = os.path.join(base_path, f"IMS{year}")
        if os.path.exists(year_folder):
            # Поиск русскоязычных текстов, исключая ключевые слова и аннотации
            text_files = [f for f in glob.glob(os.path.join(year_folder, "*_rus.txt"))
                          if not any(x in f for x in ['KW', 'Abstract'])]
            if text_files:
                file_path = text_files[0]
                text = read_text_file(file_path)
                if text:
                    selected_texts.append({
                        'year': year,
                        'file': os.path.basename(file_path),
                        'text': text
                    })

    # Анализ собранных текстов
    analysis_results = []
    for text_data in selected_texts:
        print(f"Анализируем текст из {text_data['file']}")
        results = analyze_text_complexity(text_data['text'])
        results['year'] = text_data['year']
        results['file'] = text_data['file']
        analysis_results.append(results)

    # Создание и настройка DataFrame
    df = pd.DataFrame(analysis_results)

    # Настройка отображения pandas
    pd.set_option("display.max_columns", None)
    pd.set_option("display.max_rows", None)
    pd.set_option("display.width", 150)
    pd.set_option("display.colheader_justify", "center")
    pd.set_option("display.precision", 2)

    # Словарь для перевода названий колонок на русский язык
    column_names_ru = {
        'year': 'Год',
        'file': 'Файл',
        'flesch_reading_ease': 'Индекс удобочитаемости Флеша',
        'flesch_kincaid_grade': 'Уровень Флеша-Кинкайда',
        'smog_index': 'Индекс SMOG',
        'coleman_liau_index': 'Индекс Колмана-Лиау',
        'automated_readability_index': 'Автоматический индекс удобочитаемости',
        'dale_chall_readability_score': 'Оценка удобочитаемости Дейла-Чолла',
        'difficult_words': 'Сложные слова',
        'gunning_fog': 'Индекс Ганнинга',
        'linsear_write_formula': 'Формула Линсеара',
        'lix': 'Индекс LIX',
        'rix': 'Индекс RIX',
        'kpc': 'KPC',
        'ttr': 'TTR',
        'noun_ratio': 'Доля существительных',
        'verb_ratio': 'Доля глаголов'
    }

    # Выбранные колонки для отображения
    selected_columns = [
        'year', 'file',
        'flesch_reading_ease', 'flesch_kincaid_grade', 'smog_index',
        'coleman_liau_index', 'automated_readability_index', 'dale_chall_readability_score',
        'difficult_words', 'gunning_fog', 'linsear_write_formula',
        'lix', 'rix', 'kpc', 'ttr', 'noun_ratio', 'verb_ratio'
    ]

    # Подготовка и вывод результатов
    print("\nТаблица метрик сложности по одному тексту из каждого года:\n")
    available_columns = [col for col in selected_columns if col in df.columns]
    df_display = df[available_columns].sort_values(by='year')
    df_display.columns = [column_names_ru[col] for col in df_display.columns]
    print(df_display.to_string(index=False))

if __name__ == "__main__":
    main()
```
