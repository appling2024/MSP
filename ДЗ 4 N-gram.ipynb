```python
import nltk
from nltk.corpus import ConllCorpusReader
from nltk.tag import UnigramTagger, BigramTagger, TrigramTagger

# Укажите путь и файл корпуса
corpus_path = 'path_to_russian_corpus/'  # Замените на реальный путь
corpus_file = 'ru_syntagrus-ud-train.conllu'  # Файл с разметкой

# Чтение корпуса
reader = ConllCorpusReader(corpus_path, corpus_file,
                           ['ignore', 'words', 'ignore', 'pos', 'ignore', 'ignore', 'ignore', 'ignore', 'ignore', 'ignore'],
                           encoding='utf-8')

# Получение размеченных предложений
tagged_sents = reader.tagged_sents(tagset='universal')

# Разделение данных на тренировочные и тестовые
size = int(len(tagged_sents) * 0.9)
train_sents = tagged_sents[:size]
test_sents = tagged_sents[size:]

# Создание Unigram, Bigram и Trigram теггеров
unigram_tagger = UnigramTagger(train_sents)
bigram_tagger = BigramTagger(train_sents, backoff=unigram_tagger)
trigram_tagger = TrigramTagger(train_sents, backoff=bigram_tagger)

# Пример предложения для анализа
sentence = ["Мама", "мыла", "раму"]

# Анализ
unigram_result = unigram_tagger.tag(sentence)
bigram_result = bigram_tagger.tag(sentence)
trigram_result = trigram_tagger.tag(sentence)

# Вывод результатов
print("Unigram Tagger Result:\n", unigram_result)
print("\nBigram Tagger Result:\n", bigram_result)
print("\nTrigram Tagger Result:\n", trigram_result)

# Оценка точности
print('\nUnigram accuracy:\t', unigram_tagger.accuracy(test_sents))
print('Bigram accuracy:\t', bigram_tagger.accuracy(test_sents))
print('Trigram accuracy:\t', trigram_tagger.accuracy(test_sents))
```

```python
import nltk
import pymorphy3
from nltk.util import ngrams
from collections import Counter

# Инициализация анализатора pymorphy3
morph = pymorphy3.MorphAnalyzer()

# Пример текста
text = "Отношение финнов к русским – тема старая и затертая. По этому поводу существует огромное количество мнений, анекдотов и различных материалов в СМИ. Отношения между представителями двух стран многократно обыгрывались в литературе и в кино. Помимо этого, существуют опросы общественного мнения и статистические исследования. Так как же относятся финны к русским, есть ли в этом отношении что-то особенное, и как в России относятся к Финляндии? Споры о том, как некие усредненные финны относятся к обобщенным русским ведутся постоянно, и постоянно упираются в противоречия частных мнений и отдельных терминов. Кто такой средний русский или средний финн? Это мужчина или женщина? Сколько ему должно быть лет, какое у него образование и социальное положение? Вопросов много, а единого ответа нет. Статистические данные в обобщениях не особо помогают – люди все слишком разные. Стоит провести какой-нибудь опрос или социологическое исследование, как устоявшиеся образы сразу распадаются на множество индивидуальностей, а затертые до дыр стереотипы оказываются мифами. Ко всему прочему, возникают сложности с терминологией и языковой принадлежностью. Как, например, определить процент «русскости» у татарина или у марийца, приезжающего в Финляндию? Именно поэтому логичнее и грамотнее говорить о россиянах, а не о русских. В Финляндии несколько раз проводились опросы на тему отношения к россиянам, и большая часть этих опросов касалась тех, кто приезжает в страну, как турист. Масштабные, по финским меркам, исследования показывают, что в целом в Финляндии около 85% представителей торговли считают российских туристов очень важными клиентами. В Лаппеенранте эта цифра достигает 98,5%. Позитивное отношение к российским туристам высказывают, в первую очередь, представители турбизнеса. И не только потому, что россияне повышают выручку. Вопреки расхожим представлениям, финны, работающие в туристической отрасли, дружно отмечают скромность, законопослушность и аккуратность российских туристов, причем, в сравнении с туристами из других стран. Позитивных оценок также часто удостаиваются шведы, немцы, датчане и французы. Еще одна важная деталь – положительное отношение к российским туристам наблюдается там, где их много. В Хельсинки, в Южной Карелии, в Котке и Хамине россияне воспринимаются лучше, чем в регионах, где они редкие гости."
# Токенизация текста на слова
words = text.split()

# Морфологический анализ слов
morphological_analysis = []
for word in words:
    parsed_word = morph.parse(word)[0]  # Получаем лучший вариант разбора
    morphological_analysis.append({
        'word': word,
        'lemma': parsed_word.normal_form,
        'pos': parsed_word.tag.POS,
        'morph_features': parsed_word.tag
    })

# Печать результатов морфологического анализа
print("Морфологический анализ:")
for analysis in morphological_analysis:
    print(f"Слово: {analysis['word']}, Лемма: {analysis['lemma']}, Часть речи: {analysis['pos']}, Морфологические признаки: {analysis['morph_features']}")

print('-' * 50)

# Построение N-грамм (униграмм, биграмм и триграмм)
# Создаем униграммы, биграммы и триграммы
unigrams = list(ngrams(words, 1))
bigrams = list(ngrams(words, 2))
trigrams = list(ngrams(words, 3))

# Подсчет частоты N-грамм
unigram_freq = Counter(unigrams)
bigram_freq = Counter(bigrams)
trigram_freq = Counter(trigrams)

# Печать частоты N-грамм
print("\nЧастота униграмм:")
for unigram, count in unigram_freq.items():
    print(f"{unigram}: {count}")

print("\nЧастота биграмм:")
for bigram, count in bigram_freq.items():
    print(f"{bigram}: {count}")

print("\nЧастота триграмм:")
for trigram, count in trigram_freq.items():
    print(f"{trigram}: {count}")
```
