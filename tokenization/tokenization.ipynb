```python
import nltk
nltk. download('punkt')
```

```python
with open('forbes1.txt', 'r', encoding = 'windows-1251') as f:
  my_file = f.read().replace('\n', ' ')

#sentence = '''Несмотря на глобальные экологические тенденции и попытки модной индустрии за ними следовать, в мире перерабатывается менее 1% текстильных отходов.'''
tokens = nltk.word_tokenize(my_file)
print(tokens)
print(len(tokens))

with open('forbes1_nltk.txt', 'w', encoding = 'UTF-8') as f2:
  for item in tokens:
    f2.write(item)
    f2.write('\n')
```

```python
import stanza
stanza.download('ru')
```

```python
with open('forbes1.txt', encoding = 'windows-1251') as f:
  my_file = f.read().replace('\n', ' ')

ppln = stanza.Pipeline('ru', processors='tokenize')
doc = ppln(my_file)
tokens = 0
for i, sentence in enumerate(doc.sentences):
  for token in sentence.tokens:
    #print(token.text)
    tokens = tokens + 1

#print('\nВсего токенов:', tokens)
print('\nВсего токенов коротким способом:', doc.num_tokens)

with open('forbes1_stanza.txt', 'w', encoding = 'UTF-8') as f2:
  for i, sentence in enumerate(doc.sentences):
    for token in sentence.tokens:
      f2.write(token.text)
      f2.write('\n')
```

```python
filenames = ['forbes1_nltk.txt', 'forbes1_stanza.txt', 'forbes1_Natasha.txt', 'forbes1_mosestokenizer.txt', 'forbes1_spacy.txt', 'forbes1_to-tmx.txt', 'forbes1_segtok.txt', 'forbes1_UdPipe.txt']
token_lists = []
for filename in filenames:
  with open(filename, 'r', encoding='utf-8') as f:
    tokens = f.readlines()
    token_lists.append([token.strip() for token in tokens])
```

```python
for i, tokens in enumerate(token_lists):
  print(f'Количество токенов в {filenames[i]}: {len(tokens)}')
```

```python
filenames = ['forbes1_Natasha.txt', 'forbes1_nltk.txt', 'forbes1_stanza.txt', 'forbes1_segtok.txt']
token_lists = []
for filename in filenames:
  with open(filename, 'r', encoding = 'utf-8') as f:
    tokens = f.readlines()
    token_lists.append([token.strip() for token in tokens])
```

```python
max_len = max(len(lst) for lst in token_lists)
```

```python
for i in range(max_len):
  tokens_at_position = [lst[i] if i < len(lst) else None for lst in token_lists]
  if len(set(tokens_at_position)) > 1:
    print(f'Различие в позиции {i}: {tokens_at_position}')
```
