```python
def wordform(name,token):
   with open(f'{name}.man.contexts.txt', 'r', encoding='windows-1251') as f:
       for line in f:
           items = line.strip('</?p>|</?se>|/n').split('</w>')
           for _ in items:
             index = _.find('/')
             #print(index)
             if index > 0:
              word = _[index+2:]
              if token == word:
                gr = _.find('gr=')
                print(_[gr+3:index])
            #print(item)
           print(items)
   #return index

print(wordform('apatija','Он'))
```

```python
def lemmaform(name,token):
   with open(f'{name}.man.contexts.txt', 'r', encoding='windows-1251') as f:
       for line in f:
           items = line.strip('</?p>|</?se>|/n').split('</w>')
           for _ in items:
             index = _.find('/')
             if index > 0:
              word = _[index+2:]
              if token == word:
                lex = _.find('lex=')
                print(_[lex+4:index])
            #print(item)
           print(items)

lemmaform('apatija','Предательств')
```
