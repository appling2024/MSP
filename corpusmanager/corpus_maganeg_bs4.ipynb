```python
from random import sample
from bs4 import BeautifulSoup
```

```python
with open('gsd.txt', encoding='utf-8') as f:
    corpus_raw = f.read()
```

```python
CORPUS = ''
for sent_id, sentence in enumerate(corpus_raw.split('\n\n')):
    CORPUS += f'<sent sent_id={sent_id}>'
    for line in sentence.split('\n'):
        if line.startswith('#') or '\t' not in line:
            continue
        tk_id, form, lemma, upos, xpos, feats, head, deprel, deps, misc = line.split('\t')
        gram_info = ''
        if '|' in feats:
            for grammeme in feats.split('|'):
                cath, val = grammeme.split('=')
                gram_info += f'{cath.lower()}="{val.lower()}" '
            CORPUS += f'<token id="{tk_id}" text="{form}" lemma="{lemma}" upos="{upos}" xpos="{xpos}" {gram_info} head="{head}" deprel="{deprel}" deps="{deps}" misc={misc}></token>'
        else:
            CORPUS += f'<token id="{tk_id}" text="{form}" lemma="{lemma}" upos="{upos}" xpos="{xpos}" feats="{feats}" head="{head}" deprel="{deprel}" deps="{deps}" misc={misc}></token>'
    CORPUS += '</sent>'
```

```python
CORPUS = BeautifulSoup(CORPUS, features='xml')
```

```python
def return_kwics(query:dict, n_examples:int, kwic_len:int, corpus=CORPUS):
    OUT = []
    answer = corpus.find_all('token', query)
    answer = sample(answer, n_examples if len(answer) >= n_examples else len(answer))
    if answer:
        for answer_token in answer:
            mini_kwik = []
            item_index = int(answer_token['id'])
            sentence = answer_token.parent
            for token in sentence.find_all('token'):
                if abs(int(token['id']) - item_index) <= kwic_len:
                    mini_kwik.append(token)   
        OUT.append(mini_kwik)
    return OUT
```

```python
return_kwics(query={'lemma': 'год'}, n_examples=2, kwic_len=2)
```

```python
from random import sample
from bs4 import BeautifulSoup
```

```python
with open('diskussiya.sem.om.txt', 'r', encoding='utf-8') as f:
    text = f.read()
CORPUS = BeautifulSoup(text, features='xml')
```

```python
def return_sentences_rnc(query:dict, n_examples:int, corpus=CORPUS):
    OUT = []
    answers = corpus.find_all('ana', query)
    answers = sample(answers, n_examples if len(answers) >= n_examples else len(answers))
    for answer in answers:
        OUT.append(' '.join(tk.text for tk in answer.parent.parent.find_all('w')))
    return OUT
```

```python
return_sentences_rnc(query={'gr': 'S m inan sg gen'}, n_examples=2)
```
