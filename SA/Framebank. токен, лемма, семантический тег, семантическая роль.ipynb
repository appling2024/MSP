а) Поиск контекстов по заданному токену (например, взглядом)

```python
import pandas as pd
import re

def load_data(file_path):
  data = pd.read_csv(file_path, sep='\t')
  return data

def find_by_token(data, token): #поиск по токену
  filtered_data = data[data['WordDep'] == token]
  return filtered_data

def main():
    file_path = "C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/SA/framebank_anno_ex_items.txt"
    token_to_search = input()
    print(f'Токен для поиска: {token_to_search}')

    data_token = load_data(file_path)

    results_token = find_by_token(data_token, token_to_search)

    if results_token.empty:
        print(f'Не найдено словоупотреблений с токеном: {token_to_search}')
    else:
      print(f'Найденные словоупотребления с токеном: "{token_to_search}":')
      print(results_token)
      print(f'\nВсего найдено словоупотреблений: {len(results_token)}')
if __name__ == "__main__":
    main()
```

б) Поиск контекстов по заданной лемме (например, взгляд)

```python
import re
with open("C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/SA/vljublennost_.man.contexts.txt", 'r', encoding='windows-1251') as f:
    file = f.read()

parts = file.split('<w><ana')

with open('output.txt', 'w', encoding='utf-8') as f:
    for i, part in enumerate(parts):
        f.write(f"{i}: {part}\n")

# Поиск по файлу
def search_corpus(file_path, search_lemma):
    with open(file_path, 'r', encoding='utf-8') as file:
        lines = file.readlines()

    results = []

    # Регулярное выражение для поиска леммы
    search_pattern = re.compile(rf'lex="([^"]*{search_lemma}[^"]*)"')

    # Поиск по строкам
    for index, line in enumerate(lines):
        if search_pattern.search(line):
            results.append(f"{index + 1}: {line.strip()}")

    if results:
        print("Найденные словоупотребления:")
        for result in results:
            print(result)
    else:
        print("Словоупотребления с такой леммой не найдены.")

    print(f'\nВсего найдено словоупотреблений: {len(results)}')

    
    with open('results_lemma.txt', 'w', encoding='utf-8') as f:
        for result in results:
            f.write(result + '\n')

# Запуск
file_path = 'output.txt'
search_lemma = "кукла"  # Лемму вводим без "lex=", т.к. мы ищем по всему тексту

search_corpus(file_path, search_lemma)
```

в) поиск контекстов по заданному семантическому тегу (например, r:concr) и/или комбинации тегов  (например, r:concr t:tool)

```python
def search_corpus(file_path, search_tag):
    with open(file_path, 'r', encoding='windows-1251') as file:
        lines = file.readlines()

    results = []

    for index, line in enumerate(lines):
        if search_tag in line:
            results.append(f"{index + 1}: {line.strip()}")

    if results:
        print("Найденные словоупотребления:")
        for result in results:
            print(result)
    else:
        print("Словоупотребления с таким тегом не найдены.")

    print(f'\nВсего найдено словоупотреблений: {len(results)}')

    # Сохранение в файл
    with open('results_tag.txt', 'w', encoding='windows-1251') as f:
        for result in results:  # записываем все результаты в файл
            f.write(result + '\n')


# Путь к файлу теперь задается через input
file_path = input("") #C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/SA/vljublennost_.man.contexts.txt
search_tag = input("")  #r:spec
search_corpus(file_path, search_tag)
```

г) поиск контекстов по заданному тегу семантической роли (например, framebank/framebank_roles.txt at master · olesar/framebank · GitHub)

```python
import pandas as pd
import random

def load_data(file_path):
  data = pd.read_csv(file_path, sep='\t')
  return data

def find_by_role(data, role): #поиск контекстов по заданному семантической роли (например, атрибут)
  filtered_data = data[data['Role'] == role]
  return filtered_data

def main():
    file_path = "C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/SA/framebank_anno_ex_items.txt"
    role_to_search = input()
    print(f'Роль для поиска: {role_to_search}')

    data_role = load_data(file_path)

    results_role = find_by_role(data_role, role_to_search)

    if results_role.empty:
        print(f'Не найдено словоупотреблений с ролью: {role_to_search}')
    else:
      print(f'Найденные словоупотребления с ролью "{role_to_search}":')
      print(results_role)
      print(f'\nВсего найдено словоупотреблений: {len(results_role)}')

      output_file_path = 'results_role.txt'  # Путь к итоговому файлу, можно изменить
      results_role.to_csv(output_file_path, sep='\t', index=False)  # Сохраняем результаты
      #print(f'Результаты сохранены в файл: {output_file_path}')

if __name__ == "__main__":
    main()
```
