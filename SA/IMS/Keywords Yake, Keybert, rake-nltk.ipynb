```python
# Чтение файлов
article_path = "C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/SA/IMS/Solovyev_IMS_2022.txt"
abstract_path = "C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/SA/IMS/Solovyev_IMS_2022_Abstract_rus.txt"
keywords_path = "C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/SA/IMS/Solovyev_IMS_2022_KW_rus.txt"

# Функция для чтения текста из файла
def read_text(file_path):
    with open(file_path, encoding='utf-8') as f:
        return f.read()

article_text = read_text(article_path)
abstract_text = read_text(abstract_path)
keywords_text = read_text(keywords_path)

# Проверим, что текст был успешно загружен
print(keywords_text[:300])  # Печать первых 500 символов статьи
```

```python
import spacy

# Загружаем модель для русского языка
nlp = spacy.load("ru_core_news_sm")

# Функция для лемматизации текста с помощью spaCy
def preprocess_text_spacy(text):
    doc = nlp(text)
    lemmatized_words = [token.lemma_ for token in doc if token.is_alpha and not token.is_stop]
    return ' '.join(lemmatized_words)

# Применяем предобработку с использованием spaCy
article_text_preprocessed = preprocess_text_spacy(article_text)
abstract_text_preprocessed = preprocess_text_spacy(abstract_text)
keywords_text_preprocessed = preprocess_text_spacy(keywords_text)

# Проверим результаты
print(article_text_preprocessed[:500])
```

```python
import yake

yake_keyword_extractor = yake.KeywordExtractor(lan="ru", n=1, dedupLim=0.9, top=10, features=None)

# Извлечение ключевых фраз из текста
yake_keywords = yake_keyword_extractor.extract_keywords(article_text_preprocessed)
yake_keywords = [keyword[0] for keyword in yake_keywords]  # Получаем только слова

print("YAKE Keywords:", yake_keywords)
```

```python
from keybert import KeyBERT

# Инициализация модели KeyBERT
kw_model = KeyBERT()

# Извлечение ключевых слов
keybert_keywords = kw_model.extract_keywords(article_text_preprocessed, top_n=10)

# Печать результатов
keybert_keywords = [keyword[0] for keyword in keybert_keywords]
print("KeyBERT Keywords:", keybert_keywords)
```

```python
from rake_nltk import Rake

# Инициализация RAKE
rake = Rake()

# Обработка текста
rake.extract_keywords_from_text(article_text_preprocessed)

# Извлечение 10 самых высокоранговых фраз
rake_keywords = rake.get_ranked_phrases()[:10]

# Печать результатов
print("RAKE Keywords:", rake_keywords)
```

```python
all_keywords = set(yake_keywords + keybert_keywords + rake_keywords)

# Печать объединённых ключевых слов
print("All Keywords:", all_keywords)
```

```python
auto_keywords_path = "C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/SA/IMS/keywords_results.txt"
gold_keywords_path = "C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/SA/IMS/Solovyev_IMS_2022_KW_rus.txt"

# Функция для загрузки ключевых слов из файла
def load_keywords(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        keywords = {line.strip() for line in f if line.strip()}
    return keywords

# Загружаем ключевые слова
extracted_keywords = load_keywords(auto_keywords_path)
file_keywords = load_keywords(gold_keywords_path)

# Находим совпадения
common_keywords = extracted_keywords & file_keywords

# Вычисляем метрики
precision = len(common_keywords) / len(extracted_keywords) if extracted_keywords else 0
recall = len(common_keywords) / len(file_keywords) if file_keywords else 0

# Выводим результаты
print(f"Автоматически извлечённые ключевые слова: {extracted_keywords}")
print(f"Ключевые слова из файла: {file_keywords}")
print(f"Совпадающие ключевые слова: {common_keywords}")
print(f"Точность (precision): {precision:.2f}")
print(f"Полнота (recall): {recall:.2f}")
```
