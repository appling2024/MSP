```python
import nltk
from nltk.parse import FeatureEarleyChartParser
from nltk.grammar import FeatureGrammar

grammar_path = "C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/test.fcfg"
with open(grammar_path, "r", encoding="utf-8") as file:
    grammar = FeatureGrammar.fromstring(file.read())
parser = FeatureEarleyChartParser(grammar)
test_sentence = "пятое июля 1993 года".split()
for tree in parser.parse(test_sentence):
    print(tree)
```
