RAKE (Rapid Automatic Keyword Extraction) --
разбивает текст на фразы по стоп-словам и пунктуации, а затем оценивает их значимость на основе частоты слов и их совместного появления.
1. Ключевые фразы могут быть длиннее триграмм
2. Учитывает семантическую связность фраз, а не только частотность
3. Может выявить фразы, которые не попали в топ-20, но имеют высокую значимость

```python
from __future__ import division
import operator
import nltk
```

```python
with open("punct.txt", 'r', encoding='utf-8') as file:
    punct = file.read()
punct_list = nltk.word_tokenize(punct)  # Токенизация пунктуации

with open("stop.txt", 'r', encoding='utf-8') as file:
    stop = file.read()
stop_list = [word.strip() for word in stop.split('\n') if word.strip()]  # Разбиваем строку на слова

mystop = list(set(punct_list + stop_list))  # Объединяем списки и удаляем дубликаты
print(mystop)
```

```python
def isPunct(word): # проверяет, является ли слово знаком препинания
  return len(word) == 1 and (word in string.punctuation or word in punct_list)

def isNumeric(word): # проверяет, является ли слово числом
  try:
    float(word) if '.' in word else int(word)
    return True
  except ValueError:
    return False

def isInitial(word): # проверяет, является ли слово инициалом
  return len(word) == 2 and word[1] == "."
```

```python
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
from nltk.corpus import stopwords
```

```python
# Класс RAKE
class RakeKeywordExtractor:
    def __init__(self):
        self.stopwords = set(nltk.corpus.stopwords.words('russian'))
        self.top_fraction = 3  # Топ треть фраз

    def _generate_candidate_keywords(self, sentences):
        phrase_list = []
        for sentence in sentences:
            words = map(lambda x: "|" if x in self.stopwords or x in stop_list or isNumeric(x) or isInitial(x) else x,
                        nltk.word_tokenize(sentence.lower()))
            phrase = []
            for word in words:
                if word == "|" or isPunct(word):
                    if len(phrase) > 0:
                        phrase_list.append(phrase)
                        phrase = []
                else:
                    phrase.append(word)
            if len(phrase) > 0:  # Добавляем последнюю фразу в предложении
                phrase_list.append(phrase)
        return phrase_list

    def _calculate_word_scores(self, phrase_list):
        word_freq = nltk.FreqDist()
        word_degree = nltk.FreqDist()
        for phrase in phrase_list:
            degree = len([x for x in phrase if not isNumeric(x)]) - 1
            for word in phrase:
                word_freq[word] += 1
                word_degree[word] += degree
        for word in word_freq.keys():
            word_degree[word] = word_degree[word] + word_freq[word]
        word_scores = {}
        for word in word_freq.keys():
            word_scores[word] = word_degree[word] / word_freq[word]
        return word_scores

    def _calculate_phrase_scores(self, phrase_list, word_scores):
        phrase_scores = {}
        for phrase in phrase_list:
            phrase_score = 0
            for word in phrase:
                phrase_score += word_scores[word]
            phrase_scores[" ".join(phrase)] = phrase_score
        return phrase_scores

    def extract(self, text, incl_scores=False):
        sentences = nltk.sent_tokenize(text)
        phrase_list = self._generate_candidate_keywords(sentences)
        word_scores = self._calculate_word_scores(phrase_list)
        phrase_scores = self._calculate_phrase_scores(phrase_list, word_scores)
        sorted_phrase_scores = sorted(phrase_scores.items(), key=operator.itemgetter(1), reverse=True)
        n_phrases = len(sorted_phrase_scores)
        if incl_scores:
            return sorted_phrase_scores[0:int(n_phrases/self.top_fraction)]
        else:
            return list(map(lambda x: x[0], sorted_phrase_scores[0:int(n_phrases/self.top_fraction)]))
```

```python
rake = RakeKeywordExtractor()
```

Pullenti -- извлекает именованные сущности: имена людей, организации, даты и т.д. Использует формологический и синтаксический анализ.

```python
from pullenti_wrapper.processor import Processor, GEO, ORGANIZATION, PERSON, PHONE, ADDRESS
from collections import Counter
```

```python
def extract_keywords_with_pullenti(text):
    pullenti_processor = Processor([GEO, ORGANIZATION, PERSON, PHONE, ADDRESS])  # Инициализация процессора
    keywords = []

    for line in text:  # Чтение файла построчно
        line = line.strip()  # Удаляем лишние пробелы
        if not line:  # Пропускаем пустые строки
            continue

        try:
            result = pullenti_processor(line)
            for match in result.walk():
                # Извлекаем имя и фамилию, если есть
                if hasattr(match.referent, 'firstname') and hasattr(match.referent, 'lastname'):
                    full_name = f"{match.referent.firstname} {match.referent.lastname}".lower()
                    keywords.append(full_name)

                # Добавляем остальные значимые атрибуты
                for key, value in match.referent.slots:
                    if key not in ('ISRELATIVE', 'SEX', 'NUMBER', 'HIGHER', 'PROFILE',
                                 'ALPHA2', 'POINTER', 'FROM', 'REF', 'ATTRIBUTE'):
                        keywords.append(str(value).lower())
        except ValueError:
            continue

    # Подсчёт частот и сохранение в CSV
    freq_word = Counter(keywords)
    top_20 = freq_word.most_common(20)

    with open('pullenti_keywords.csv', 'w', encoding='utf-8') as f:
        for term, count in top_20:
            f.write(f"{term};{count}\n")

    print(top_20)
```

```python
with open('100heroes.txt', 'r', encoding='windows-1251') as file:
    extract_keywords_with_pullenti(file)
```
