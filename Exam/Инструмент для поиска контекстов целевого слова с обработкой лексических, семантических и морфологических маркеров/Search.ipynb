```python
import random

FRAMEBANK_FILE = 'framebank_anno_ex_items.txt'
```

```python
# Посмотреть первые 5 строк корпуса
with open(FRAMEBANK_FILE, 'r', encoding='utf-8') as f:
    for _ in range(6):
        print(f.readline().strip())
```

```python
# Поиск по слову без учёта роли
def search_by_word(file, word, sample_size=10):
    results = []
    word = word.lower()
    with open(file, 'r', encoding='utf-8') as f:
        next(f)  # пропускаем заголовок
        for line in f:
            items = line.strip().split('\t')
            if word in items[4].lower() or word in items[5].lower():
                results.append(line.strip())
    if results:
        sample = random.sample(results, min(sample_size, len(results)))
        print('ConstrIndex\tExIndex\tPlace\tPhraseGen\tWordDep\tForm\tRole\tRank\tSem\tRea\tConstrExId\tItemExIndex\tKeyLexemes')
        print('\n'.join(sample))
    else:
        print('совпадений не найдено')

search_by_word(FRAMEBANK_FILE, 'алкать')
```

```python
def search_by_word_and_rank(file, word, rank, sample_size=10):
    results = []
    word = word.lower()
    rank = rank.lower()
    with open(file, 'r', encoding='utf-8') as f:
        next(f)
        for line in f:
            items = line.strip().split('\t')
            if (word in items[4].lower() or word in items[5].lower()) and rank in items[7].lower():
                results.append(line.strip())
    if results:
        sample = random.sample(results, min(sample_size, len(results)))
        print('ConstrIndex\tExIndex\tPlace\tPhraseGen\tWordDep\tForm\tRole\tRank\tSem\tRea\tConstrExId\tItemExIndex\tKeyLexemes')
        print('\n'.join(sample))
    else:
        print('совпадений не найдено')

search_by_word_and_rank(FRAMEBANK_FILE, 'алкать', 'предикат')
```

```python
# Поиск только по роли
def search_by_role(file, role, sample_size=10):
    results = []
    role = role.lower()
    with open(file, 'r', encoding='utf-8') as f:
        next(f)
        for line in f:
            items = line.strip().split('\t')
            if role in items[6].lower():
                results.append(line.strip())
    if results:
        sample = random.sample(results, min(sample_size, len(results)))
        print('ConstrIndex\tExIndex\tPlace\tPhraseGen\tWordDep\tForm\tRole\tRank\tSem\tRea\tConstrExId\tItemExIndex\tKeyLexemes')
        print('\n'.join(sample))
    else:
        print('совпадений не найдено')

search_by_role(FRAMEBANK_FILE, 'субъект')
```
