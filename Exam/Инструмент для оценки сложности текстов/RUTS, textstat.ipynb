Немного о метриках:
Flesch Reading Ease - Рассчитывается через среднюю длину предложений и среднее число слогов на слово. Оценивает читаемость текста по шкале от 0 до 100 (может быть больше 100 и меньше 0): чем выше, тем текст легче.

SMOG - то же самое, что и Gunning Fog (различие: Gunning Fog - берется средняя длина предложения и сложные слова в предложении, SMOG - квадратный корень сложных слов на предложение) 

Coleman-Liau Index - Учитывает количество букв вместо слогов – упрощает автоматический расчет (среднее число букв на 100 слов и среднее число предложений на 100 слов). Результат – примерный класс образования: чем выше, тем сложнее


```python
from pathlib import Path
import pandas as pd
from ruts import ReadabilityStats
import textstat
```

```python
# тексты
folder_path = Path('/Users/sergey/Library/Mobile Documents/com~apple~CloudDocs/Учеба СПбГУ/Семантические анализаторы/Экзамен/Оценка сложности текстов/10 Texts')

# Считываем все файлы в словарь
texts = {}
for filepath in folder_path.glob("*.txt"):
    with open(filepath, encoding="utf-8") as f:
        texts[filepath.name] = f.read()
```

RUTS

```python
rows = []

for filename, text in texts.items():
    rs = ReadabilityStats(text)
    stats = rs.get_stats()
    stats["Файл"] = filename
    rows.append(stats)

# Создание DataFrame и перестановка колонок
df_ruts = pd.DataFrame(rows)
df_ruts = df_ruts[["Файл", 
                   "flesch_kincaid_grade",
                   "flesch_reading_easy",
                   "coleman_liau_index",
                   "smog_index",
                   "automated_readability_index",
                   "lix"]]

df_ruts
```

textstat

```python
textstat_results = []

for filename, text in texts.items():
    textstat_results.append({
        "Файл": filename,
        "Flesch Reading Ease": textstat.flesch_reading_ease(text),
        "SMOG Index": textstat.smog_index(text),
        "Flesch-Kincaid Grade": textstat.flesch_kincaid_grade(text),
        "Coleman-Liau Index": textstat.coleman_liau_index(text),
        "Automated Readability Index": textstat.automated_readability_index(text),
        "Dale-Chall Score": textstat.dale_chall_readability_score(text)
    })

df_textstat = pd.DataFrame(textstat_results)
df_textstat
```
