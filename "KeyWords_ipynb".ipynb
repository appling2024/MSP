```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
with open('Belkin_IMS_2024_Abstract_rus.txt', 'r', encoding = 'utf-8') as f:
  abstr = f.read()

with open('Belkin_IMS_2024_rus.txt', 'r', encoding = 'utf-8') as file:
  artcl = file.read()
```

**Yake**

```python
!pip install git+https://github.com/LIAAD/yake
```

```python
import yake
```

```python
language = 'ru'
max_ngram_size = 2
numOfKeywords = 10
windowSize = 1

custom_kw_extractor = yake.KeywordExtractor(lan=language, n=max_ngram_size, windowsSize=windowSize, top=numOfKeywords, features=None)
keywords = custom_kw_extractor.extract_keywords(abstr)

for kw in keywords:
  print(kw)

#неправильно, ниже другой способ
embedding = model.encode(kw)

similarities = model.similarity(embedding, embedding)
print(similarities)
```

```python
sentences1_yake = ['Пользовательский фидбек',
'частности поликлиниками',
'проблемных аспектов',
'фидбек является',
'является ценным',
'ценным источником',
'источником информации',
'пользовательских отзывов',
'отзывов',
'оказываемых медицинскими']
```

```python
embedding = model.encode(sentences1_yake)
print(embedding.shape) #вывод покажет, что у нас есть 10 векторов, каждый из которых состоит из 384 значений
similarities = model.similarity(embedding, embedding) #вычисление матрицы сходства: где значение == 1Б там элемент равен сам себе, то есть полностью схож сам с собой
print(similarities)
```

```python
import keybert
from keybert import KeyBERT
```

```python
kw_model = KeyBERT()
```

```python
keywords_abs = kw_model.extract_keywords(abstr, keyphrase_ngram_range=(1, 2), top_n = 10)
keywords_abs
```

```python
sentences_abstr_bert = ['источником информации',
 'медицинских учреждениях',
 'медицинскими учреждениями',
 'ценным источником',
 'тематического моделирования',
 'повышения удовлетворенности',
 'поликлиниками выявления',
 'оказываемых медицинскими',
 'отзывов медицинских',
 'информации']
```

```python
embedding = model.encode(sentences_abstr_bert)
print(embedding.shape)
similarities = model.similarity(embedding, embedding)
print(similarities)
```

```python
keywords_art = kw_model.extract_keywords(artcl, keyphrase_ngram_range=(1, 2), top_n = 10)
keywords_art
```

```python
sentence_artcl_bert = ['взаимодействия предотвращение',
 'привлечения пользователей',
 'медучреждениям пользователи',
 'пользователей информацию',
 'сообщения некоторых',
 'стремительного развития',
 'сообщениях разнообразные',
 'что информация',
 'пользователями учреждениям',
 'автоматического определения']
```
