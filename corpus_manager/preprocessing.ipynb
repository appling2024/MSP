```python
import pandas as pd
```

```python
df = pd.read_csv('framebank_anno_ex_circ.txt', sep='\t')
```

```python
df = df[['Phrase', 'Form', 'Role', 'Type', 'KeyLexemes']]
```

```python
DATA = []
for row in df.values.tolist():
    for key_lexeme in row[4].split(','):
        DATA.append([row[0].lower(), row[1], row[2], row[3], key_lexeme.lower()])
```

```python
df_clear = pd.DataFrame(DATA, columns=df.columns)
df_clear.drop_duplicates(inplace=True)
df_clear
```

```python
df_clear.to_csv('framebank.csv', index=False)
```
