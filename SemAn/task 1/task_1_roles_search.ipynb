```python
roles_list = []
with open('roles.txt', 'r') as file:
  for line in file:
    line = line.strip('\n')
    if line != '' and '#' not in line:
      roles_list.append(line)
```

```python
def search(role):
  x = 0
  list_mentioned = []
  with open('framebank_anno.txt', 'r') as f:
    for line in f:
      if x > 0:
        line1 = line.split('\t')
        if line1[5] == role and line1[3] not in list_mentioned:
        #if line1[5] == role:
          print(line1[3])
          list_mentioned.append(line1[3])
      x += 1
    if list_mentioned == []:
      if role not in roles_list:
        print('В системе нет такой семантической роли')
      else:
        print('Нет примеров для такой семантической роли')

search('возможность')
```
