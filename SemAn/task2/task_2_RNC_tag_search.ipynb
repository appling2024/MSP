```python
import re

words = []
sent = []
tags = []
tags_full = []

def lines_func(name):
  with open(f'{name}.txt','r',encoding='windows-1251') as f:

    lines = []
    #убираем лишние символы
    for line in f:
      line = re.sub('<\/?w?>|\\t|<|\n','',line)
      lines.append(line)

    #создаем списки слов и тэгов
    for line in lines:
      words = []
      tags = []
      line = line.split('ana ')
      for i,elem in enumerate(line):
        if i == 0:
          continue
        else:
          line1 = elem.split('/>')
          tags.append(line1[0])
          words.append(line1[-1])
      tags_full.append(tags)
      sent.append(words)

def context(tag):
  index = dict()
  #достаем индексы из списка тэгов
  for i,tags in enumerate(tags_full):
    for j,t in enumerate(tags):
      if tag in t:
        index[i] = j

  #по индексам находим слово и контекст
  for i,words in enumerate(sent):
    for j,word in enumerate(words):
      if j == index.get(i):
        print()
        print(word, tag)
        string = ''
        for k in words:
          string += k
        print(string)

lines_func('britva')
context('r:concr t:tool:instr d:dim der:s')
```
