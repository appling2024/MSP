```python
with open('ChizhikMelnikova_IMS_2022_Abstract_rus.txt','r') as f:
  text_abstract = f.read()

with open('ChizhikMelnikova_IMS_2022.txt','r') as f:
  text_full = f.read()
```

```python
import yake

# assuming default parameters
kw_extractor = yake.KeywordExtractor()
keywords = kw_extractor.extract_keywords(text_abstract)

for kw in keywords:
    print(kw)
```

```python
with open('YAKE_top_20_abstract.csv', 'w', encoding='windows-1251') as f:
    for eachkwrd in keywords:
        f.write(str(eachkwrd[0])+";"+str(eachkwrd[1])+"\n")
```

```python
import keybert
from keybert import KeyBERT
```

```python
with open('ChizhikMelnikova_IMS_2022.txt', 'r', encoding='utf-8') as f:
    doc_full = f.read()
```

```python
kw_model = KeyBERT()
keywords = kw_model.extract_keywords(doc_full, top_n=20)
keywords
```

```python
with open('KeyBERT_top_20.csv', 'w', encoding='windows-1251') as f2:
    for eachkwrd in keywords:
        f2.write(str(eachkwrd[0])+";"+str(eachkwrd[1])+"\n")
```
