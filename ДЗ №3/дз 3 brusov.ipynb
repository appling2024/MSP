```python
import spacy
from pymorphy3 import MorphAnalyzer

nlp_spacy = spacy.load("ru_core_news_sm")

morph = MorphAnalyzer()

file_path = "C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/ДЗ №3/brusov.txt"

with open(file_path, "r", encoding="windows-1251") as file:
    text = file.read()


# 1. Анализ с помощью SpaCy
doc_spacy = nlp_spacy(text)
spacy_results = []
for token in doc_spacy:
    spacy_results.append(f"{token.text}\t{token.lemma_}\t{token.pos_}\t{token.morph}\n")

with open("C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/ДЗ №3/spacy_results.txt", "w", encoding="utf-8") as spacy_file:
    spacy_file.writelines(spacy_results)

# 2. Анализ с помощью pymorphy3
pymorphy_results = []
for word in text.split():
    parsed = morph.parse(word)[0]
    pymorphy_results.append(f"{word}\t{parsed.normal_form}\t{parsed.tag}\n")

with open("C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/ДЗ №3/pymorphy3_results.txt", "w", encoding="utf-8") as pymorphy_file:
    pymorphy_file.writelines(pymorphy_results)

print("Анализ завершён. Результаты сохранены в файлы spacy_results.txt и pymorphy3_results.txt.")
```
