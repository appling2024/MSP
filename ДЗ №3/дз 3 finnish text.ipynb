```python
import spacy
import stanza

nlp_spacy = spacy.load("fi_core_news_sm")

stanza.download("fi") 
nlp_stanza = stanza.Pipeline("fi")

def analyze_with_spacy(input_file, output_file):
    with open("C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/ДЗ №3/Minun paras ystäväni.txt", "r", encoding="utf-8") as file:
        text = file.read()
    doc = nlp_spacy(text)
    with open(output_file, "w", encoding="utf-8") as file:
        file.write("Слово\tЛемма\tЧасть речи\tСинтаксис\tПризнаки\n")
        for token in doc:
            file.write(f"{token.text}\t{token.lemma_}\t{token.pos_}\t{token.dep_}\t{token.morph}\n")

def analyze_with_stanza(input_file, output_file):
    with open(input_file, "r", encoding="utf-8") as file:
        text = file.read()
    doc = nlp_stanza(text)
    with open(output_file, "w", encoding="utf-8") as file:
        file.write("Слово\tЛемма\tЧасть речи\tXPOS\tПризнаки\n")
        for sentence in doc.sentences:
            for word in sentence.words:
                file.write(f"{word.text}\t{word.lemma}\t{word.upos}\t{word.xpos}\t{word.feats}\n")

input_filename = "C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/ДЗ №3/Minun paras ystäväni.txt"
spacy_output_filename = "C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/ДЗ №3/spacy_finnish_analysis.txt"
stanza_output_filename = "C:/Users/Asus/OneDrive/Рабочий стол/Библиотеки/MSP/ДЗ №3/stanza_finnish_analysis.txt"

analyze_with_spacy(input_filename, spacy_output_filename)
analyze_with_stanza(input_filename, stanza_output_filename)

print(f"Результаты анализа сохранены в файлы:\n- {spacy_output_filename}\n- {stanza_output_filename}")
```
