```python
import re

# удалить ьеги 
def clean_text(line):
    return re.sub(r"<[^>]+>", "", line).strip()


def search_lemma(file_path, lemma):
    results = []
    pattern = re.compile(rf"lex='{lemma}'")  
    with open(file_path, "r", encoding="windows-1251") as f:
        for line in f:
            if pattern.search(line):  
                clean_line = clean_text(line)  
                results.append((line.strip(), clean_line))
    return results

def search_pos(file_path, pos):
    results = []
    pattern = re.compile(rf"gr='{pos},|gr='{pos}'")  
    with open(file_path, "r", encoding="windows-1251") as f:
        for line in f:
            if pattern.search(line):  
                clean_line = clean_text(line)  
                results.append((line.strip(), clean_line))
    return results


def search_deps(file_path, dep):
    results = []
    with open(file_path, "r", encoding="windows-1251") as f:
        for line in f:
            if dep in line and "sem=" in line: 
                clean_line = clean_text(line)  
                results.append((line.strip(), clean_line))
    return results


file_path = "lom.txt" 
lemma = "лом" 
pos = "S"  
dep = "t:tool:instr"  


lemma_results = search_lemma(file_path, lemma)
print(f"Найдено {len(lemma_results)} совпадений для леммы '{lemma}':")
for idx, (raw, clean) in enumerate(lemma_results[:10], start=1):  
    print(f"{idx}. Исходная строка с тегами:\n{raw}\nЧистый текст:\n{clean}\n")
print("— — — —" * 10)  


pos_results = search_pos(file_path, pos)
print(f"\nНайдено {len(pos_results)} совпадений для части речи '{pos}':")
for idx, (raw, clean) in enumerate(pos_results[:10], start=1):  
    print(f"{idx}. Исходная строка с тегами:\n{raw}\nЧистый текст:\n{clean}\n")
print("— — — —" * 10)  


dep_results = search_deps(file_path, dep)
print(f"\nНайдено {len(dep_results)} совпадений для зависимости '{dep}':")
for idx, (raw, clean) in enumerate(dep_results[:10], start=1): 
    print(f"{idx}. Исходная строка с тегами:\n{raw}\nЧистый текст:\n{clean}\n")
```
