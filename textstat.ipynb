```python
import textstat

def analyze_text(text: str) -> dict:
    return {
        "Количество символов": textstat.char_count(text),
        "Количество слов": textstat.lexicon_count(text, removepunct=True),
        "Количество предложений": textstat.sentence_count(text),
        "Легкость чтения": textstat.flesch_reading_ease(text), #оценивает, насколько легко читать текст: чем выше результат, тем проще текст
        "Индекс SMOG": textstat.smog_index(text), #какой уровень образования требуется для того, чтоб прочитать текст
        "Индекс удобочитаемости": textstat.flesch_kincaid_grade(text), #мера определения сложности восприятия текста читателем. Индекс удобочитаемости может вычисляться на основе нескольких параметров: длины предложений, слов, удельного количества наиболее частотных (или редких) слов и т. д.
        "Индекс Колман — Лиау": textstat.coleman_liau_index(text), #сложность восприятия текста читателем путём аппроксимирования сложности текста к номеру класса в американской системе образования, ученикам которого данный текст будет понятен
        "Трудность понимания при чтении текста": textstat.dale_chall_readability_score(text),
        "Автоматизированный индекс читаемости (ARI)": textstat.automated_readability_index(text),
        "Количество сложных слов": textstat.difficult_words(text),
        "Средняя оценка уровняя образования, необходимого для прочтения текста": textstat.text_standard(text),
    }
```

```python
def main():
    print('Оценка сложности текста \n')
    results = analyze_text(text)

    for metric, value in results.items():
        print(f"{metric}: {value}")

if __name__ == "__main__":
    main()
```
