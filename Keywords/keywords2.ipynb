```python
import pandas as pd
```

```python
import keybert
from keybert import KeyBERT
```

```python
with open('/Users/sergey/PycharmProjects/SemAnalyzer/Keywords/ChizhikMelnikovaZakharov_IMS_2022.txt', 'r', encoding='utf-8') as f:
    doc = f.read()
```

```python
with open('/Users/sergey/PycharmProjects/SemAnalyzer/Keywords/ChizhikMelnikovaZakharov_IMS_2022_Abstract_rus.txt', 'r', encoding='utf-8') as f:
    doc_abstract = f.read()
```

```python
kw_model = KeyBERT()
```

```python
keywords = kw_model.extract_keywords(doc)
```

```python
keywords = kw_model.extract_keywords(doc, top_n=20)
```

```python
with open('KeyBERT_top_20_bigram.csv', 'w', encoding='utf-8') as f2:
    for eachkwrd in keywords:
        f2.write(str(eachkwrd[0])+";"+str(eachkwrd[1])+"\n")
```

```python
keywords = kw_model.extract_keywords(doc, keyphrase_ngram_range=(1, 3), top_n=20)
```

```python
with open('KeyBERT_top_20_trigram.csv', 'w', encoding='utf-8') as f2:
    for eachkwrd in keywords:
        f2.write(str(eachkwrd[0])+";"+str(eachkwrd[1])+"\n")
```

```python
#KeyBERT Abstract
```

```python
keywords = kw_model.extract_keywords(doc_abstract, top_n=20)
with open('KeyBERT_abstract_bigram.csv', 'w', encoding='utf-8') as f2:
    for eachkwrd in keywords:
        f2.write(str(eachkwrd[0])+";"+str(eachkwrd[1])+"\n")
```

```python
keywords = kw_model.extract_keywords(doc_abstract, keyphrase_ngram_range=(1, 3), top_n=20)
with open('KeyBERT_abstract_trigram.csv', 'w', encoding='utf-8') as f2:
    for eachkwrd in keywords:
        f2.write(str(eachkwrd[0])+";"+str(eachkwrd[1])+"\n")
```

```python
df1 = pd.read_csv('KeyBERT_top_20_bigram.csv', sep=';', header=None)[0]
df2 = pd.read_csv('KeyBERT_abstract_bigram.csv', sep=';', header=None)[0]
df3= pd.read_csv('KeyBERT_top_20_trigram.csv', sep=';', header=None)[0]
df4= pd.read_csv('KeyBERT_abstract_trigram.csv', sep=';', header=None)[0]

pd.concat([df1, df2, df3, df4], axis=1)
```

```python
from rutermextract import TermExtractor
```

```python
term_extractor = TermExtractor()
```

```python
terms = term_extractor(doc, limit=20)
```

```python
with open('RuTermExtract_top_20.csv', 'w', encoding='utf-8') as f2:
    for term in terms:
        f2.write(f"{term.normalized};{term.count}\n")
```

```python
#rutermAbstract
```

```python
terms_sec = term_extractor(doc_abstract, limit=20)
with open('RuTermExtract_abstract.csv', 'w', encoding='utf-8') as f2:
    for term in terms_sec:
        f2.write(f"{term.normalized};{term.count}\n")
```

```python
df1 = pd.read_csv('RuTermExtract_top_20.csv', sep=';', header=None)[0]
df2 = pd.read_csv('RuTermExtract_abstract.csv', sep=';', header=None)[0]

pd.concat([df1, df2], axis=1)
```
